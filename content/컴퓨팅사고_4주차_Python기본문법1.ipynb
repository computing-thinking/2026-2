## 1-2. 타입(type)이란? — `type()` 으로 확인하기

사람은 `7`, `7.0`, `"7"` 을 모두 ‘칠’이라고 읽지만, 컴퓨터에게는 **전혀 다른 종류의 값**입니다.
값의 종류를 **데이터 타입(data type, 자료형)** 이라고 하며, 타입에 따라 **할 수 있는 연산**이 달라집니다.

| 타입 | 이름 | 예 | 쓰임 |
|:---:|:---|:---|:---|
| `int` | 정수 (integer) | `10`, `-5`, `0` | 개수, 순서, 나이 |

| `str` | 문자열 (string) | `"sea"`, `'123'` | 이름, 문장, 학번 |
| `bool` | 불리언 (boolean) | `True`, `False` | 조건 판단의 결과 |

내장 함수 `type()` 에 값을 넣으면 타입을 알려 줍니다.

### 진법(base)에 따른 정수 표현

2주차에 본 것처럼 컴퓨터 내부는 2진수입니다. Python에서는 앞에 붙이는 **접두어**로 진법을 표시합니다.

| 접두어 | 진법 | 예 | 10진수 값 |
|:---:|:---|:---|:---:|
| (없음) | 10진수 | `10` | 10 |
| `0b` | 2진수 (binary) | `0b1010` | 10 |
| `0o` | 8진수 (octal) | `0o12` | 10 |
| `0x` | 16진수 (hexadecimal) | `0xA` | 10 |

````python
print(0b1010)    # 2진수 1010
print(0o12)      # 8진수 12
print(0xA)       # 16진수 A
````

````python
# 반대로: 10진수 → 다른 진법 (결과는 문자열로 나옵니다)
print(bin(10))
print(oct(10))
print(hex(255))
````

## 1-4. 실수형 `float`

소수점이 있는 수입니다. 두 가지 방법으로 적을 수 있습니다.

| 표기법 | 예 | 특징 |
|:---|:---|:---|
| 소수점 표기 | `3.14`, `-2.5` | 직관적 |
| 지수 표기 (`e`) | `1.0e3`, `1.0e-3` | 아주 크거나 아주 작은 수를 간결하게 |

`314159e-5` 는 **314159 × 10⁻⁵** 이라는 뜻입니다. `314159` 를 가수(유효숫자), `-5` 를 지수라고 합니다.

````python
print(1.0e3)        # 1.0 × 10³
print(1.0e-3)       # 1.0 × 10⁻³
print(314159e-5)
````

````python
avogadro = 6.022e23       # 아보가드로 수
print(avogadro)
print(type(avogadro))
````

### 실수를 저장하는 두 가지 방법 — 고정소수점과 부동소수점

컴퓨터는 수를 **정해진 개수의 칸**에 저장해야 합니다. 쉽게 설명하려고, 10진수 숫자 **8칸**만 쓸 수 있다고 해 봅시다.

**① 고정소수점 (fixed point)** — 소수점 위치를 미리 정해 둡니다. 예: 앞 4칸은 정수 부분, 뒤 4칸은 소수 부분.

<span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#E6F7F6;color:#0B7C7B;border:1px solid #0EA5A4;">1</span><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#E6F7F6;color:#0B7C7B;border:1px solid #0EA5A4;">2</span><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#E6F7F6;color:#0B7C7B;border:1px solid #0EA5A4;">3</span><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#E6F7F6;color:#0B7C7B;border:1px solid #0EA5A4;">4</span><b style="font-size:1.3em;padding:0 2px;">.</b><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#FFF1EF;color:#C2410C;border:1px solid #FF6B5A;">5</span><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#FFF1EF;color:#C2410C;border:1px solid #FF6B5A;">6</span><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#FFF1EF;color:#C2410C;border:1px solid #FF6B5A;">7</span><span style="display:inline-block;min-width:1.5em;padding:2px 0;text-align:center;font-family:monospace;font-weight:bold;border-radius:4px;margin:1px;background:#FFF1EF;color:#C2410C;border:1px solid #FF6B5A;">8</span> &nbsp; → `1234.5678`

- 표현할 수 있는 범위: `0000.0001` ~ `9999.9999`
- `12345.6` (너무 큼), `0.00001` (너무 작음) 은 저장할 수 없습니다.

**② 부동소수점 (floating point)** — 칸을 **가수**와 **지수**로 나눕니다. 위에서 본 `e` 표기와 같은 방식입니다. 예: 6칸은 가수, 2칸은 지수.

## 1-7. 글자는 어떻게 저장될까? — 문자 인코딩

컴퓨터의 메모리에는 0과 1, 즉 **숫자만** 저장할 수 있습니다. 그래서 글자마다 **번호를 약속**해 두고 그 번호를 저장합니다. 이 약속을 **문자 인코딩(character encoding)** 이라고 합니다.

### ASCII (아스키, 1963)

- **A**merican **S**tandard **C**ode for **I**nformation **I**nterchange
- 한 글자를 **7비트**(0~127)로 표현 → 영어 알파벳, 숫자, 기호, 제어문자 128자
- 제어문자(0~31): 줄바꿈(LF, 10), ESC(27) 등 화면에 찍히지 않는 문자 · 인쇄 가능 문자(32~126): 공백, 숫자, 알파벳, 기호
- 한글은 표현할 수 없습니다.

**ASCII 표 (일부)**

| 10진수 | 16진수 | 2진수 (7비트) | 문자 |
|---:|---:|:---:|:---:|
| 10 | 0x0A | `0001010` | LF (줄바꿈) |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 32 | 0x20 | `0100000` | 공백 |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 48 | 0x30 | `0110000` | `0` |
| 49 | 0x31 | `0110001` | `1` |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 57 | 0x39 | `0111001` | `9` |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 65 | 0x41 | `1000001` | `A` |
| 66 | 0x42 | `1000010` | `B` |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 90 | 0x5A | `1011010` | `Z` |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 97 | 0x61 | `1100001` | `a` |
| 98 | 0x62 | `1100010` | `b` |
| ⋮ | ⋮ | ⋮ | ⋮ |
| 122 | 0x7A | `1111010` | `z` |

- 글자 `'0'` 은 0번이 아니라 **48번**입니다. `'0'` ~ `'9'` 는 48 ~ 57번에 차례로 있습니다.
- 대문자와 소문자는 번호가 정확히 **32** 차이 납니다 (`'A'` = 65, `'a'` = 97).
- 알파벳과 숫자는 번호가 **순서대로** 붙어 있어서 크기 비교와 계산을 할 수 있습니다.

`ord()` 는 글자 → 번호, `chr()` 는 번호 → 글자입니다.

````python
print(ord("A"))
print(ord("a"))
print(chr(65))
print(chr(65 + 1))      # 번호니까 계산도 됩니다
````

### 유니코드(Unicode)와 UTF-8

- **유니코드**: 전 세계 모든 문자에 고유 번호(**코드 포인트**)를 붙인 문자 집합. `U+` 뒤에 16진수로 적습니다. 예) `'A'` = U+0041, `'가'` = U+AC00
- 번호 공간은 `0x0 ~ 0x10FFFF` (약 111만 개)이며, 한글 · 한자 · 이모지까지 모두 들어 있습니다.
- **UTF-8**: 유니코드 번호를 **1~4바이트의 가변 길이**로 저장하는 방식. ASCII 문자는 1바이트 그대로여서 기존 영어 문서와 호환됩니다. 오늘날 웹의 표준입니다.

#### 틀 (x 자리에 번호의 비트를 채움)

| 번호 범위 | 바이트 수 | 비트 틀 | 담을 수 있는 비트 |
|:---|:---:|:---|:---:|
| U+0000 ~ U+007F | 1 | `0xxxxxxx` | 7 |
| U+0080 ~ U+07FF | 2 | `110xxxxx 10xxxxxx` | 11 |
| U+0800 ~ U+FFFF | 3 | `1110xxxx 10xxxxxx 10xxxxxx` | 16 |
| U+10000 ~ U+10FFFF | 4 | `11110xxx 10xxxxxx 10xxxxxx 10xxxxxx` | 21 |

- **첫 바이트**의 앞부분 `0`, `110`, `1110`, `11110` 이 이 글자가 몇 바이트인지 알려 줍니다.
- **이어지는 바이트**는 항상 `10` 으로 시작합니다.

**예: `가` (U+AC00)**

1. 번호를 2진수로 바꿉니다: `AC00` = `1010 1100 0000 0000` (16비트)
2. 16비트이므로 3바이트 틀을 쓰고, 4 · 6 · 6비트로 나눕니다: `1010` | `110000` | `000000`
3. 틀에 채웁니다:

```
1110 1010   10 110000   10 000000
   E    A      B    0      8    0     →  EA B0 80
```

`"가".encode("utf-8")` 의 결과가 `b'\xea\xb0\x80'` 인 이유가 이것입니다.

````python
print(ord("가"))
print(hex(ord("가")))        # U+AC00
print(chr(0xAC00 + 1))       # '가' 다음 글자는?
````

````python
# encode() : 문자열을 실제 저장되는 바이트로 변환
print("A".encode("utf-8"),  "→", len("A".encode("utf-8")),  "바이트")
print("가".encode("utf-8"), "→", len("가".encode("utf-8")), "바이트")
print("😀".encode("utf-8"), "→", len("😀".encode("utf-8")), "바이트")
````

### 한글 인코딩 — 한글이 ‘깨지는’ 이유

| | EUC-KR | CP949 | 유니코드 / UTF-8 |
|:---|:---|:---|:---|
| 한글 글자 수 | 2,350자 (완성형) | 11,172자 | 11,172자 (U+AC00 ~ U+D7A3) |
| 한글 1자 크기 | 2바이트 | 2바이트 | 3바이트 |
| `'가'` 저장 값 | `B0 A1` | `B0 A1` | `EA B0 80` |
| 한글 순서 | 가나다순 | 추가된 글자는 순서가 어긋남 | 완전한 가나다순 |
| 표현할 수 있는 문자 | 한국어 + 일부 한자 | 한국어 + 일부 한자 | 전 세계 문자 |
| 쓰인 시기 | 1980~2000년대 한글 문서의 주류 | Windows 95 이후, 지금도 한국어 Windows 기본값 | 2010년대 중반 이후 표준 |

- **CP949**는 마이크로소프트가 EUC-KR에 빠진 한글을 더해 만든 방식입니다. EUC-KR에 있던 글자는 저장 값이 똑같습니다.
- Windows 의 메모장·엑셀 등에서 인코딩 이름이 **ANSI** 로 표시되면, 한국어 Windows 에서는 바로 **CP949** 를 뜻합니다. (메뉴의 ‘유니코드’ 는 UTF-16 을 뜻하므로 UTF-8 과 헷갈리지 마세요.)
- **유니코드**는 한글 11,172자를 **가나다 순서대로 빠짐없이** 배치했습니다. 앞에서 `chr(0xAC00 + 1)` 이 `'각'` 을 돌려준 것도 이 덕분입니다.

같은 글자도 **약속이 다르면 저장되는 숫자가 다릅니다.**

````python
print("가".encode("utf-8"))
print("가".encode("euc-kr"))
````

````python
# UTF-8 로 저장한 글을 CP949 약속으로 읽으면?
# errors="replace" : 해석할 수 없는 바이트를 만나면 오류 대신 � 로 바꿔 표시
print("가".encode("utf-8").decode("cp949", errors="replace"))
saved = "안녕, 세상".encode("utf-8")
print(saved.decode("cp949", errors="replace"))

# 반대로, CP949 로 저장한 '가'(B0 A1)를 UTF-8 약속으로 읽으면?
print(b"\xb0\xa1".decode("utf-8", errors="replace"))
````

### 📎 숫자와 문자열에서 사용하는 접두사

<details><summary><b>▶ 클릭해서 펼쳐 보기</b> — <code>0x</code>, <code>\x</code>, <code>U+</code> 가 헷갈릴 때</summary>

| 분류 | 쓰는 자리 | 결과 |
|:---|:---|:---|
| ① 숫자 접두사 | 코드에서 **숫자 앞** | `int` 값 |
| ② 이스케이프 | **따옴표 안** | 글자(`str`) 또는 바이트(`bytes`) |
| ③ 문자열 접두사 | **따옴표 앞** | 문자열의 종류를 정함 |
| ④ `U+` | **코드가 아님**. 문서와 표에서 쓰는 표기 | 없음 (Python 문법이 아님) |

#### ① 숫자 접두사 — 정수를 몇 진법으로 적었는지

| 표기 | 진법 | 예 | 값 |
|:---|:---|:---|:---|
| (없음) | 10진수 | `65` | 65 |
| `0x` / `0X` | 16진수 | `0x41` | 65 |
| `0b` / `0B` | 2진수 | `0b1000001` | 65 |
| `0o` / `0O` | 8진수 | `0o101` | 65 |

- 대문자와 소문자 모두 됩니다. 보통은 소문자로 씁니다.
- 결과는 모두 **같은 `int` 65**입니다. 적는 방법만 다릅니다.
- 반대로 숫자를 이런 표기의 문자열로 바꾸는 함수가 있습니다: `hex(65)` → `'0x41'`, `bin(65)` → `'0b1000001'`, `oct(65)` → `'0o101'`

#### ② 이스케이프 — 따옴표 안에서 글자나 바이트를 번호로 적기

| 표기 | 자릿수 | `str`에서 | `bytes`에서 |
|:---|:---|:---|:---|
| `\x` | 16진수 2자리 | `'\x41'` → `'A'` | `b'\x41'` → 바이트 65 |
| `\u` | 16진수 4자리 | `'\uac00'` → `'가'` | 사용 불가 |
| `\U` | 16진수 8자리 | `'\U0001F600'` → `'😀'` | 사용 불가 |
| `\N{…}` | 유니코드 이름 | `'\N{HANGUL SYLLABLE GA}'` → `'가'` | 사용 불가 |
| `\n`, `\t` | 정해진 기호 | 줄바꿈, 탭 | 줄바꿈, 탭 |

- `\u`와 `\U`는 **대소문자에 따라 뜻이 다릅니다**(4자리와 8자리). `\x`는 반드시 소문자로 씁니다.

#### ③ 문자열 접두사 — 따옴표 앞 글자

| 표기 | 뜻 | 예 |
|:---|:---|:---|
| `b'…'` | 바이트열(`bytes`) | `b'A'` |
| `f'…'` | f-string (값 끼워 넣기, 2교시에 배웁니다) | `f'{x}원'` |
| `r'…'` | raw: 이스케이프 무시 | `r'C:\new'` → `\n`이 줄바꿈이 되지 않음 |
| `u'…'` | 유니코드 문자열. Python 3에서는 **아무 효과가 없음** (Python 2 시절의 흔적) | `u'가'` == `'가'` |

#### ④ `U+` — 유니코드 표준의 문서 표기

- `U+AC00`처럼 **유니코드 번호를 사람에게 알려 줄 때** 쓰는 표기입니다. 표, 문서, 웹사이트에 나옵니다.
- **Python 코드가 아닙니다.** `U+AC00`을 코드에 쓰면 오류가 납니다.
- 규칙: 대문자 `U+` 뒤에 **대문자 16진수를 최소 4자리** 씁니다. 예: `U+0041`, `U+AC00`, `U+1F600`
- **`u+`는 표준 표기가 아닙니다.** 가끔 보이는 것은 잘못 적었거나 비공식으로 쓴 것입니다.
- 코드로 옮길 때는 형식을 바꿔 적습니다.
  - `U+AC00` → 숫자로는 `0xAC00`, 문자열로는 `'\uac00'`
  - 반대로 번호를 `U+` 형식으로 출력하려면 `f"U+{ord('가'):04X}"` → `'U+AC00'`

#### 한 번에 보기 — 전부 `'가'`, 44032를 가리킴

```python
44032                     # 10진수 정수
0xAC00                    # 16진수로 적은 같은 정수          ① 숫자 접두사

'\uac00'                  # '가'                            ② 이스케이프
'가'.encode('utf-8')      # b'\xea\xb0\x80'  ← UTF-8 바이트  ②③ bytes 안의 \x
# U+AC00                  ← 문서에서 이 글자를 가리키는 이름    ④ (코드 아님)
```

**간단히 외우는 법**

- `0x` : **숫자**를 16진수로 적을 때
- `\x`, `\u` : **따옴표 안**에서 글자를 번호로 적을 때
- `U+` : **사람이 읽는 문서**에서 유니코드 번호를 가리킬 때

## 2-2. `=` 은 ‘같다’가 아니라 ‘넣어라’

### 컴퓨팅 사고 — 콜라 컵과 사이다 컵의 내용물을 맞바꾸려면?

컵이 두 개뿐이면 바꿀 수 없습니다. **빈 컵이 하나 더** 필요합니다. 변수도 같습니다.
아래 코드는 **잘못된** 방법입니다. 왜 그런지 예상해 보고 실행하세요.

````python
cup1 = "콜라"
cup2 = "사이다"

cup1 = cup2      # cup1 에 사이다를 부으면… 콜라는?
cup2 = cup1
print(cup1, cup2)
````

올바른 방법 — 빈 컵(`temp`) 사용:

````python
cup1 = "콜라"
cup2 = "사이다"

temp = cup1      # ① 콜라를 빈 컵에 옮겨 두고
cup1 = cup2      # ② cup1 에 사이다를 붓고
cup2 = temp      # ③ 빈 컵의 콜라를 cup2 에
print(cup1, cup2)
````

Python에서는 한 줄로도 쓸 수 있습니다. (오른쪽 값들을 **먼저 모두 꺼낸 뒤** 왼쪽에 넣기 때문에 가능합니다.)

````python
cup1, cup2 = "콜라", "사이다"
cup1, cup2 = cup2, cup1
print(cup1, cup2)
````

## 2-7. 출력 — `print()` 와 f-string

`print()` 에는 값을 쉼표로 여러 개 넣을 수 있습니다. 사이에 공백이 하나씩 들어갑니다.

````python
name = "김연세"
score = 95
print(name, "학생의 점수는", score, "점입니다.")
````

문장 안에 변수 값을 끼워 넣을 때는 **f-string** 이 가장 편합니다. 따옴표 **앞에 `f`** 를 붙이고, 값을 넣을 자리에 **`{변수}`** 를 씁니다.

````python
print(f"{name} 학생의 점수는 {score}점입니다.")
print(f"10점을 더하면 {score + 10}점")          # { } 안에는 식도 쓸 수 있습니다
````

`{값:형식}` 으로 표시 형식을 정할 수 있습니다. 자주 쓰는 것 두 가지만 기억하세요.

````python
pi = 3.14159265
price = 1234567
print(f"원주율은 약 {pi:.2f}")        # .2f → 소수 둘째 자리까지
print(f"가격: {price:,}원")           # ,   → 세 자리마다 쉼표
````

## 3-1. 식(expression) — 계산하면 ‘값’이 되는 것

**식**은 Python이 계산해서 **하나의 값으로 바꿀 수 있는 모든 것**입니다.

| 식의 종류 | 예 | 값 |
|:---|:---|:---|
| 리터럴 | `5` | 5 |
| 변수 | `score` | 변수에 든 값 |
| 연산자로 결합한 것 | `5 + 6`, `3 < 2` | 11, False |
| 함수 호출 | `len("hello")` | 5 |
| 위의 것들을 조합한 것 | `len("hello") * 2 + score` | … |

노트북이 셀 마지막 줄의 값을 보여 주는 것은, 그 줄이 **식**이기 때문입니다.

````python
len("hello") * 2 + 1
````

## 3-2. 산술 연산자

| 연산자 | 뜻 | 예 | 결과 |
|:---:|:---|:---|:---:|
| `+` | 덧셈 | `7 + 2` | 9 |
| `-` | 뺄셈 | `7 - 2` | 5 |
| `*` | 곱셈 | `7 * 2` | 14 |
| `/` | 나눗셈 (결과는 **항상 실수**) | `7 / 2` | 3.5 |
| `//` | **몫** (정수 나눗셈) | `7 // 2` | 3 |
| `%` | **나머지** | `7 % 2` | 1 |
| `**` | 거듭제곱 | `7 ** 2` | 49 |

연산자 앞뒤의 공백은 결과에 영향을 주지 않습니다. 읽기 좋게 한 칸씩 띄우는 것이 관례입니다.

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
````

### `//` 와 `%` 는 어디에 쓰나? — 생각보다 아주 많이 씁니다

| 상황 | 식 |
|:---|:---|
| 짝수/홀수 판별 | `n % 2` 가 0이면 짝수 |
| 초 → 분, 초 | `t // 60`, `t % 60` |
| 끝자리 숫자 뽑기 | `n % 10` |
| 끝자리 떼어 내기 | `n // 10` |
| 30명을 4명씩 조 편성 | `30 // 4` 조, `30 % 4` 명 남음 |

````python
total = 3725                     # 3725초는 몇 시간 몇 분 몇 초?
hour = total // 3600
minute = total % 3600 // 60
second = total % 60
print(f"{total}초 = {hour}시간 {minute}분 {second}초")
````

## 3-3. 연산 결과의 타입 — 자동(암묵적) 변환

| 연산 | 결과 타입 | 예 |
|:---|:---:|:---|
| 정수 ∘ 정수 (`/` 제외) | `int` | `2 + 3` → `5` |
| 정수 `/` 정수 | `float` | `6 / 3` → `2.0` |
| 실수 ∘ 실수 | `float` | `2.1 + 3.2` |
| 정수 ∘ 실수 | `float` | `2 + 3.0` → `5.0` |

정수와 실수가 섞이면 Python이 **정수를 자동으로 실수로 바꿔서** 계산합니다. 이를 **암묵적 형 변환(implicit type coercion)** 이라고 합니다. 정보가 손실되지 않는 방향(정수 → 실수)으로만 자동 변환됩니다.

````python
print(2 + 3,   type(2 + 3))
print(2 + 3.0, type(2 + 3.0))
print(2 * 0.5, type(2 * 0.5))
````

## 3-4. 명시적 형 변환 — 내가 직접 바꾸기

자동으로 바뀌지 않는 경우에는 **변환 함수**로 직접 바꿉니다 (explicit type coercion).

| 함수 | 역할 | 예 |
|:---|:---|:---|
| `int(x)` | 정수로 | `int(3.9)` → `3`, `int("12")` → `12` |
| `float(x)` | 실수로 | `float("3.5")` → `3.5`, `float(2)` → `2.0` |
| `str(x)` | 문자열로 | `str(-12.5)` → `"-12.5"` |
| `bool(x)` | 참/거짓으로 | `bool(0)` → `False`, `bool(7)` → `True` |

````python
print(int(6 / 3))        # 2.0 → 2
print(str(6 < 3))        # False → "False"
print(float("3.5") * 2)
````

**예상해 보기** — `int(3.9)` 는 3일까요, 4일까요?

## 3-5. 비교 연산자와 논리 연산자 — 결과가 `bool` 인 식

| 비교 연산자 | 뜻 | | 논리 연산자 | 뜻 |

## 3-6. 연산자 우선순위

한 식에 연산자가 여러 개 있으면 **우선순위가 높은 것부터** 계산합니다. 수학의 ‘곱셈 먼저’와 같은 원리입니다.

| 순위 | 연산자 | 설명 | 같은 순위끼리는 |
|:---:|:---|:---|:---|
| 1 | `( )` | 괄호 | — |
| 2 | `**` | 거듭제곱 | **오른쪽 → 왼쪽** |
| 3 | `+x`, `-x` | 부호 (단항) | — |
| 4 | `*`, `/`, `//`, `%` | 곱셈 · 나눗셈 · 몫 · 나머지 | 왼쪽 → 오른쪽 |

| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=` | 비교 | — |
| 7 | `not` → `and` → `or` | 논리 | 왼쪽 → 오른쪽 |
| 마지막 | `=` | 대입 — 오른쪽 식을 **모두 계산한 뒤** 넣는다 | — |

## 3-7. 명령문(statement) — 컴퓨터에게 시키는 ‘문장’

**식**이 ‘값’이라면, **명령문**은 ‘**~해라**’ 라는 **하나의 실행 단위**입니다. 프로그램은 명령문을 **위에서 아래로 차례로** 실행합니다 — 3주차의 **순차 구조**, 그리고 폰 노이만 기계의 **명령형(imperative)** 사고방식 그대로입니다.

| 명령문 | 예 | 하는 일 |
|:---|:---|:---|
| 대입문 | `a = 7` | 값을 변수에 넣는다 |
| 식 문장 | `print(a)` | 식을 계산(함수를 실행)한다 |
| `if` 문 | `if a > 5:` | 조건에 따라 실행을 **선택**한다 |
| `while` / `for` 문 | `while a > 0:` | 실행을 **반복**한다 |
| `import` 문 | `import keyword` | 다른 사람이 만든 기능을 가져온다 |

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 식 vs 명령문</b><br><code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">a + 1</code> 은 식 (값 8 이 된다). <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">b = a + 1</code> 은 명령문 (값을 계산해서 b 에 넣는 <b>동작</b>을 한다). 명령문 안에는 식이 재료로 들어갑니다.</div>

````python
a = 7                                  # 대입문
print(a)                               # 식 문장 (함수 호출)
if a > 5:                              # if 문 — 다음 주에 자세히!
    print("a is greater than 5")       #   들여쓰기 된 줄은 조건이 참일 때만 실행
````

### 📎 참고 — C 언어에서는 변수를 어떻게 쓸까?

Python 은 변수를 **선언 없이** 바로 만들고, 담는 값에 따라 타입이 **바뀔 수도** 있습니다(2-4). C 언어는 정반대입니다. 많은 언어가 C 방식을 따르므로 한 번 비교해 두면 좋습니다.

```c
int    age    = 20;        // 정수
double height = 1.75;      // 실수
char   grade  = 'A';       // 글자 하나 (작은따옴표)
char   name[20] = "김연세"; // 문자열은 글자들의 '배열'
```

- **미리 선언해야 합니다.** 변수를 쓰기 전에 `int`, `double` 처럼 **타입을 먼저 적습니다.**
- **타입이 고정됩니다.** `age` 는 끝까지 정수입니다. `age = "스물";` 이라고 쓰면 실행 전에 오류가 납니다.
- **크기가 정해져 있습니다.** `int` 는 보통 4바이트라서 약 ±21억까지만 담깁니다. 넘으면 값이 엉뚱하게 바뀝니다. Python 의 `int` 는 한계가 없습니다.
- **문자열 타입이 따로 없습니다.** 글자(`char`)들을 배열에 늘어놓고, 끝에 널 문자(`'\0'`)를 넣어 "여기가 끝"을 표시합니다(1-7의 인코딩 이야기와 이어집니다).
- **참/거짓도 원래 없었습니다.** 0 을 거짓, 0 이 아닌 값을 참으로 썼습니다.

| | Python | C |
|:---|:---|:---|
| 선언 | 필요 없음 (`age = 20`) | 필요함 (`int age = 20;`) |
| 타입 | 값에 따라 정해지고 바뀔 수 있음 | 미리 정하면 끝까지 고정 |
| 정수 범위 | 제한 없음 | 타입에 따라 정해짐 (`int` 는 약 ±21억) |
| 문자열 | `str` 타입 | `char` 배열 + 널 문자 |
| 오류를 만나는 때 | **실행할 때** | **실행 전(번역할 때)** |
| 문장 끝 | 줄바꿈 | 세미콜론 `;` |
