````python
#@title ▶ 먼저 실행하세요 — 자가 점검 도구 준비
import importlib.util as _ilu

def _in_class_site():
    # 수업 사이트(JupyterLite)에서 열었는지 확인합니다. Google Colab 이면 False 입니다.
    return _ilu.find_spec("google") is None or _ilu.find_spec("google.colab") is None

if not _in_class_site():
    print("⚠️ 이 노트북은 Google Colab 용이 아닙니다.")
    print("   LearnUs 에 올린 수업 사이트 링크로 열어 주세요. 여기서는 자가 점검이 동작하지 않습니다.")
    raise SystemExit

def check(name, test):
    """실습 답을 확인하는 도구입니다. 수정하지 마세요."""
    if not _in_class_site():
        print(f"⚠️ {name}: 수업 사이트에서 실행해야 점검할 수 있습니다.")
        return
    try:
        ok = test()
    except NameError as e:
        print(f"⏳ {name}: 아직 필요한 변수가 없습니다 → {e}")
        return
    except Exception as e:
        print(f"❌ {name}: 확인 중 오류가 났습니다 → {type(e).__name__}: {e}")
        return
    print(f"✅ {name}: 통과!" if ok else f"❌ {name}: 값이 다릅니다. 다시 확인해 보세요.")

# 수업 사이트에서는 %%writefile 로 방금 만든 .py 파일을 import 가 바로 찾지 못합니다 (파일 목록 캐시).
# import 때마다 캐시를 비우는 작은 장치를 끼워 둡니다.
import importlib as _il, sys as _sys

class _RefreshFinder:
    _busy = False                     # invalidate_caches 가 안에서 다시 import 를 일으켜도 되돌아 들어오지 않게

    @classmethod
    def find_spec(cls, name, path=None, target=None):
        if not cls._busy:
            cls._busy = True
            try:
                _il.invalidate_caches()
            finally:
                cls._busy = False
        return None

if not any(type(f).__name__ == "_RefreshFinder" for f in _sys.meta_path):
    _sys.meta_path.insert(0, _RefreshFinder())

print("준비 완료! 이제 수업을 시작합니다.")
````

## 1-2. 파일에 쓰기 — `open` · `write` · `close`

파일을 다루는 일은 늘 세 단계입니다. **열고**, **읽거나 쓰고**, **닫습니다.** 왜 꼭 열어야 하고, 왜 꼭 닫아야 할까요? 그림으로 봅니다.

<div style="text-align:center;margin:10px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><div style="color:#6B7280;font-size:0.9em;margin-top:4px;">파일을 열고 · 쓰고 · 닫는 세 단계. 프로그램(메모리)과 디스크의 파일은 떨어져 있어서, 그 사이에 <b>통로</b>가 있어야 오갈 수 있습니다.</div></div>

**① 열기 = 통로 만들기.** 프로그램은 메모리에 있고 파일은 디스크에 있습니다. 둘은 떨어져 있습니다. `open()` 은 그 사이에 **통로**를 놓고 이름표(`f`)를 붙입니다. 전화를 거는 것과 같습니다. 통화가 연결되어야 말을 주고받을 수 있듯, 통로가 있어야 읽고 쓸 수 있습니다. 그래서 읽든 쓰든 **먼저 열어야** 합니다.

**② 쓰기 = 버퍼에 쌓기.** 디스크는 메모리보다 수천 배 느립니다. 그래서 `write()` 는 디스크에 바로 쓰지 않고 통로 안의 **버퍼**(buffer, 임시 바구니)에 모아 둡니다. 이 순간 디스크의 파일은 **아직 비어 있습니다.** 택배를 부치면 트럭이 바로 떠나지 않고 상자가 찰 때까지 기다리는 것과 같습니다.

**③ 닫기 = 버퍼 비우고 통로 끊기.** `close()` 는 버퍼에 남은 것을 디스크로 **보내고**, 통로를 걷고, 이름표 `f` 를 쓸 수 없게 합니다. 그제야 파일에 내용이 있습니다. 닫지 않으면 두 가지 문제가 생깁니다. 버퍼에 남은 내용이 디스크에 **못 갈 수** 있고, 통로가 열려 있는 동안 다른 프로그램이 그 파일을 **못 열** 수 있습니다. 전화를 끊지 않으면 마지막 말이 전달됐는지 알 수 없고, 상대도 다른 전화를 못 받는 것과 같습니다.

**예상해 보기** — `write` 는 무엇을 돌려줄까요? 그리고 출력에는 몇 줄이 보일까요?

````python
f = open("memo.txt", "w", encoding="utf-8")     # ① 열기 — "w" 는 쓰기 모드. 없으면 만들고, 있으면 비웁니다
n = f.write("첫 줄입니다\n")                       # ② 쓰기 — 줄바꿈 \n 은 직접 넣어야 합니다
f.write("둘째 줄입니다\n")
f.close()                                        # ③ 닫기

print(n)
print(os.listdir("."))
````

`write` 는 **쓴 글자 수**를 돌려줍니다 (`\n` 도 한 글자). 그리고 파일 목록에 `memo.txt` 가 새로 생겼습니다. 메뉴의 `File → Open…` 에서도 보입니다. 열어 보면 두 줄이 들어 있습니다.

`print` 와 달리 `write` 는 **줄바꿈을 넣어 주지 않습니다.** `\n` 을 빼면 두 줄이 한 줄로 이어집니다. 숫자도 바로 쓸 수 없습니다. `f.write(42)` 는 `TypeError` 입니다. f-string 으로 문자열을 만들어 씁니다.

<div style="background:#F3F4F6;color:#1F2937;border-left:6px solid #6B7280;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#6B7280;">📎 print 도 사실은 write 입니다 — 표준 출력</b><br>그런데 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">print</code> 와 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">write</code> 는 남남이 아닙니다. 프로그램은 시작할 때 운영체제가 미리 열어 둔 파일 하나를 받습니다. 바로 <b>표준 출력</b>(standard output), Python 에서는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sys.stdout</code> 입니다. 이 파일의 반대쪽 끝에는 디스크가 아니라 <b>화면</b>이 붙어 있습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">print(&quot;안녕&quot;)</code> 은 사실 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sys.stdout.write(&quot;안녕\n&quot;)</code> 을 대신 해 주는 함수입니다. 값을 문자열로 바꾸고, 사이에 띄어쓰기를 넣고, 끝에 줄바꿈을 붙여서 그 파일에 <b>씁니다.</b> 그래서 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">print</code> 에는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">file=</code> 인자가 있습니다. 화면 대신 다른 파일을 주면 거기에 씁니다.<br>이렇게 보면 화면도, 디스크의 파일도 <b>쓰는 방법이 같습니다.</b> 열고 닫는 수고가 없을 뿐입니다. 표준 출력은 프로그램이 시작할 때 이미 열려 있고 끝날 때 운영체제가 닫아 줍니다. 노트북에서는 그 ‘화면’이 셀 아래의 출력 칸입니다.</div>

**예상해 보기** — 아래 셀을 실행하면 화면에 몇 줄이 보이고, `memo2.txt` 에는 무엇이 들어갈까요?

````python
import sys

print(sys.stdout)                               # 표준 출력 — 미리 열려 있는 파일 객체. 반대쪽 끝은 화면
sys.stdout.write("print 없이 화면에 씁니다\n")        # write 로 화면에 쓰기 — 줄바꿈은 직접
print("print 는 write 에 줄바꿈을 붙인 것")        # 같은 파일에 씁니다

with open("memo2.txt", "w", encoding="utf-8") as f:
    print("이번엔 화면이 아니라 파일로", 2026, file=f)      # file= 에 파일을 주면 print 가 그 파일에 씁니다

print(open("memo2.txt", encoding="utf-8").read())
````

`open` 의 두 번째 인자가 **모드**(mode)입니다.

| 모드 | 뜻 | 파일이 이미 있으면 | 파일이 없으면 |
|:---:|:---|:---|:---|
| `"r"` | 읽기 (기본값 — 생략 가능) | 처음부터 읽는다 | **오류** (`FileNotFoundError`) |
| `"w"` | 새로 쓰기 | **내용을 지우고** 새로 쓴다 | 만든다 |
| `"a"` | 덧붙이기 (append) | 끝에 이어 쓴다 | 만든다 |

**예상해 보기** — 아래 셀을 실행하면 `memo.txt` 는 몇 줄이 될까요? 모드를 `"w"` 로 바꾸면요?

````python
f = open("memo.txt", "a", encoding="utf-8")     # "a" — 끝에 덧붙입니다
f.write("셋째 줄 — 덧붙였습니다\n")
f.close()

print(os.path.getsize("memo.txt"), "바이트")      # 파일 크기 (글자 수가 아닙니다 — 한글은 한 글자가 3바이트)
````

### `with` 문 — 닫는 것을 잊지 않게

그림의 ②를 직접 확인해 봅니다. 닫기 **전에** 파일 크기를 재면 몇 바이트일까요?

**예상해 보기** — 두 `getsize` 는 각각 무엇을 출력할까요?

````python
g = open("buffer_test.txt", "w", encoding="utf-8")
g.write("아직 안 닫았습니다")
print("닫기 전:", os.path.getsize("buffer_test.txt"), "바이트")     # 버퍼에만 있고 디스크에는 아직 없습니다

g.close()
print("닫은 뒤:", os.path.getsize("buffer_test.txt"), "바이트")     # close() 가 버퍼를 디스크로 보냈습니다
````

닫기 전에는 **0 바이트**입니다. `write()` 가 분명히 실행됐는데 디스크의 파일은 비어 있습니다. 내용은 버퍼에 있습니다. `close()` 뒤에야 디스크에 갑니다. `close()` 를 잊은 채 프로그램이 끝나면 Python 이 대개 대신 닫아 주지만, 프로그램이 도중에 죽거나 전원이 꺼지면 버퍼의 내용은 **영영 사라집니다.**

그래서 Python 은 **블록이 끝나면 자동으로 닫아 주는** 문법을 만들었습니다. `with` 문입니다. 앞으로 파일은 **항상 이 모양**으로 엽니다.

````python
with open("memo.txt", "a", encoding="utf-8") as f:      # as 뒤의 이름이 파일 객체입니다
    f.write("넷째 줄 — with 문으로 썼습니다\n")
                                                        # 들여쓰기가 끝나면 자동으로 close() 됩니다
print("닫혔나요?", f.closed)
````

<span style="background:#FFF0EE;color:#FF6B5A;border:1px solid #FF6B5A;padding:2px 10px;border-radius:999px;font-weight:bold;">💥 일부러 오류를 내는 셀 — 닫힌 파일에 쓰면?</span>

````python
f.write("다섯째 줄\n")          # with 블록 밖 — f 는 이미 닫혔습니다
````

`ValueError: I/O operation on closed file` — “닫힌 파일에 입출력을 하려 했다”는 안내입니다. 그림의 ③처럼 통로가 걷혀 이름표 `f` 로는 더 쓸 수 없습니다. 파일 작업은 **`with` 블록 안에서만** 합니다.

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 파일 쓰기 정리</b><br>① 파일 작업은 <b>열기 → 쓰기 → 닫기</b>. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">with open(...) as f:</code> 로 열면 블록 끝에서 <b>자동으로 닫힙니다.</b><br>② 모드: <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">&quot;w&quot;</code> 는 지우고 새로, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">&quot;a&quot;</code> 는 끝에 덧붙이기. 읽기는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">&quot;r&quot;</code> 이고 생략할 수 있습니다.<br>③ <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">write</code> 는 <b>문자열만</b> 받고 <b>줄바꿈을 넣지 않습니다.</b> 숫자는 f-string 으로, 줄 끝에는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">\n</code> 을 직접.<br>④ <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">encoding=&quot;utf-8&quot;</code> 은 항상 적습니다 (1-4 에서 이유를 봅니다).</div>

## 1-3. 파일 읽기 — `read` · 줄 단위 `for` · `readlines`

읽기는 모드를 생략합니다. 읽는 방법은 세 가지입니다. 먼저 **통째로** 읽습니다.

**예상해 보기** — `repr(text)` 는 어떻게 보일까요? `print(text)` 와 무엇이 다를까요?

````python
with open("memo.txt", encoding="utf-8") as f:
    text = f.read()                  # 파일 전체를 문자열 하나로

print(repr(text))                    # 있는 그대로 — 줄바꿈이 \n 으로 보입니다
print("----")
print(text)                          # 보통 출력 — \n 이 줄바꿈으로 바뀝니다
print(type(text), len(text), "글자")
````

6px;border-radius:4px;border:1px solid #BFE9E7;">memo.txt</code> 를 한 줄씩 읽은 것과 같은 동작입니다. 노트북에서는 키보드 대신 셀 아래에 입력 칸이 뜨지만, 그 칸이 곧 표준 입력입니다.<br>오류가 <b>빨간 글씨</b>로 보이는 이유도 여기 있습니다. 오류 메시지는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sys.stdout</code> 이 아니라 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sys.stderr</code> 라는 <b>다른 파일</b>로 가고, 노트북이 그 파일의 내용을 빨갛게 보여 줍니다. 키보드 · 화면 · 디스크의 파일을 모두 ‘파일’이라는 <b>하나의 개념</b>으로 읽고 쓰는 것, 이것이 운영체제가 파일을 설계한 방식입니다.</div>

**예상해 보기** — 두 `print` 는 어디로 갈까요? 둘 다 화면에 보일까요?

````python
import sys

print(sys.stdin)                                   # 표준 입력 — 반대쪽 끝은 키보드(노트북에서는 입력 칸)
print("이 줄은 표준 출력으로 갑니다")
print("이 줄은 표준 오류로 갑니다", file=sys.stderr)   # 같은 화면이지만 다른 파일 — 빨갛게 보입니다
````

데이터 파일은 보통 **한 줄이 한 건**입니다. 그래서 줄 단위로 읽는 일이 가장 많습니다. 파일 객체는 6주차의 이터레이터처럼 `for` 에 바로 넣을 수 있습니다.

**예상해 보기** — 아래 셀의 출력은 왜 줄 사이가 비어 보일까요?

````python
with open("memo.txt", encoding="utf-8") as f:
    for line in f:                   # 한 줄씩 — 메모리에 한 줄만 올리므로 큰 파일도 됩니다
        print(line)
````

`line` 에는 줄 끝의 `\n` 이 **포함되어** 있고, `print` 가 **또** 줄바꿈을 넣기 때문입니다. 두 가지 해결이 있습니다. `print(line, end="")` 로 `print` 의 줄바꿈을 끄거나, `line.strip()` 으로 양 끝의 공백과 줄바꿈을 **떼어 냅니다.** 데이터를 다룰 때는 거의 언제나 `strip()` 입니다. 줄 끝의 `\n` 이 숫자나 이름에 붙어 따라다니면 안 되기 때문입니다.

````python
with open("memo.txt", encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):       # 6주차 enumerate — 몇 번째 줄인지 함께
        print(number, repr(line.strip()))
````

세 번째 방법은 줄들을 **리스트로** 받는 것입니다. 5주차의 문자열 메서드 `splitlines()` 를 쓰면 `\n` 이 떨어진 깨끗한 리스트가 됩니다.

````python
with open("memo.txt", encoding="utf-8") as f:
    lines = f.read().splitlines()    # 줄 단위 리스트 — \n 은 떼어 줍니다

print(len(lines), "줄")
print(lines)
print(lines[-1])                     # 마지막 줄
````

**예상해 보기** — 같은 파일을 한 번 열어 **두 번** 읽으면 두 번째는 무엇이 나올까요?

````python
with open("memo.txt", encoding="utf-8") as f:
    first = f.read()
    second = f.read()

print(len(first), "글자 /", len(second), "글자")
````

두 번째는 **빈 문자열**입니다. 파일에는 ‘지금 어디까지 읽었는가’를 가리키는 위치가 있고, 한 번 끝까지 읽으면 그 위치가 끝에 가 있습니다. 책갈피가 마지막 장에 꽂힌 것입니다. 다시 읽으려면 **다시 엽니다.** 보통은 한 번 읽어 변수에 담아 두고 그 변수를 씁니다.

## 1-4. 인코딩 — 왜 `encoding="utf-8"` 을 늘 적는가

4주차에서 글자는 숫자(코드)로, 숫자는 바이트로 저장된다고 했습니다. 그 규칙이 **문자 인코딩**이고, 오늘날의 표준은 **UTF-8** 입니다. 파일은 바이트이므로, 파일을 열 때 Python 은 “이 바이트를 어떤 규칙으로 글자로 바꿀까”를 알아야 합니다.

`encoding` 을 적지 않으면 **운영체제의 기본값**을 씁니다. 수업 사이트 · 리눅스 · 맥은 UTF-8 입니다. 그런데 **한국어 Windows 는 `cp949`** 라는 옛 규칙일 수 있습니다. 같은 코드가 내 노트북에서는 되고 친구 노트북에서는 깨지는 이유가 대개 이것입니다.

**예상해 보기** — UTF-8 로 쓴 `memo.txt` 를 `cp949` 규칙으로 읽으면 어떻게 될까요?

<span style="background:#FFF0EE;color:#FF6B5A;border:1px solid #FF6B5A;padding:2px 10px;border-radius:999px;font-weight:bold;">💥 일부러 오류를 내는 셀 — 인코딩을 잘못 짚으면?</span>

````python
with open("memo.txt", encoding="cp949") as f:    # UTF-8 로 쓴 파일을 cp949 규칙으로 읽어 봅니다
    print(f.read())
````

`UnicodeDecodeError: 'cp949' codec can't decode byte ...` — “이 바이트를 cp949 규칙으로는 글자로 못 바꾸겠다”는 안내입니다. 운이 나쁘면 오류 대신 **엉뚱한 글자**(`ìë`, `안녕` 같은 것)가 나오기도 합니다. 4주차에서 본 ‘한글 깨짐’이 바로 이것입니다.

대처는 간단합니다. **쓸 때도 읽을 때도 `encoding="utf-8"` 을 적습니다.** 그러면 어느 운영체제에서 열어도 같습니다.

<div style="background:#F3F4F6;color:#1F2937;border-left:6px solid #6B7280;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#6B7280;">📎 Windows 에서 받은 파일이 깨질 때</b><br>메모장이나 엑셀(CSV 저장)에서 만든 파일은 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">cp949</code> 일 수 있습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">encoding=&quot;utf-8&quot;</code> 로 열어 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">UnicodeDecodeError</code> 가 나면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">encoding=&quot;cp949&quot;</code> 로 바꿔 보세요. 반대로 그 파일을 UTF-8 로 다시 저장하는 것이 더 좋은 해결입니다. 메모장의 ‘다른 이름으로 저장’에 인코딩 선택이 있습니다.</div>

## 2-2. `try` · `except` — 해 보고, 안 되면

“**해 보고**, 오류가 나면 **이렇게 하라**”를 적는 문법입니다.

```python
try:
    오류가 날 수 있는 코드
except 오류종류:
    오류가 났을 때 할 일
그 뒤의 코드   ← 오류가 났든 안 났든 계속 실행됩니다
```

**예상해 보기** — 이번에는 몇 줄이 출력될까요? `text` 를 `"42"` 로 바꾸면요?

````python
text = "23.5"
print("변환을 시작합니다")
try:
    n = int(text)
    print("정수:", n)                              # 오류가 나면 이 줄은 건너뜁니다
except ValueError:
    print(f"정수가 아닙니다: {text!r}")            # !r 은 repr — 따옴표를 붙여 보여 줍니다
print("프로그램 끝")                               # 어느 쪽이든 여기는 실행됩니다
````

흐름은 두 갈래입니다. `try` 블록을 위에서부터 한 줄씩 실행합니다. **끝까지 오류가 없으면** `except` 블록은 실행되지 않고 그냥 지나갑니다. **도중에 오류가 나면** 그 줄에서 `try` 블록을 빠져나와(남은 줄은 실행되지 않습니다) `except` 블록을 실행합니다. 두 경우 모두 `except` 블록 다음 줄인 `print("프로그램 끝")` 으로 이어집니다. 오류가 났는데도 프로그램이 **멈추지 않았습니다.**

오류 객체를 `as 이름` 으로 받으면 Python 의 안내문을 그대로 쓸 수 있습니다. 이것을 함수로 묶어 여러 입력에 써 봅니다.

````python
def to_int(text):
    """문자열을 정수로. 안 되면 None 을 돌려준다."""
    try:
        return int(text)
    except ValueError as e:          # e 가 오류 객체 — 안내문이 들어 있습니다
        print(f"  (건너뜀: {e})")
        return None

for text in ["42", " 7 ", "23.5", "스물", ""]:
    print(repr(text), "→", to_int(text))
````

`int()` 는 앞뒤 공백은 봐주지만 `"23.5"` · `"스물"` · 빈 문자열은 거부합니다. 전에는 첫 번째 거부에서 프로그램이 멈췄지만, 이제는 **다섯 개를 다 처리**합니다. 안 되는 것은 `None` 으로 표시하고 지나갑니다.

## 2-3. 오류 종류별로 다르게 — `except` 여러 개

한 블록에서 여러 종류의 오류가 날 수 있습니다. `except` 를 여러 개 두면 **종류별로** 다르게 대처합니다.

**예상해 보기** — 세 입력 각각 어느 `except` 로 갈까요?

````python
menu = ["김밥", "라면", "떡볶이"]

for choice in ["1", "5", "둘"]:
    try:
        index = int(choice)                 # "둘" 이면 여기서 ValueError
        print(choice, "→", menu[index])     # 5 면 여기서 IndexError
    except ValueError:
        print(choice, "→ 숫자로 골라 주세요")
    except IndexError:
        print(choice, "→ 0 부터", len(menu) - 1, "사이로 골라 주세요")
````

`except` 는 위에서부터 **종류가 맞는 첫 번째** 것이 실행됩니다. 모든 오류를 한꺼번에 받으려면 맨 뒤에 `except Exception as e:` 를 둡니다. `Exception` 은 모든 예외의 부모 클래스입니다(9주차 상속). 다만 **어떤 오류인지 모른 채 삼키면** 나중에 원인을 찾기 어렵습니다. 종류를 아는 오류는 종류로 잡고, 마지막 그물은 안내문을 **출력**합니다.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("  0 으로 나눌 수 없습니다")
    except Exception as e:                           # 나머지 전부 — 마지막 그물
        print(f"  예상 못 한 오류: {type(e).__name__}: {e}")
    return None

print(safe_divide(10, 4))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
````

## 2-4. `raise` — 내가 오류를 내다

지금까지는 Python 이 내는 오류를 **받았습니다.** 반대로 **내가 오류를 낼** 수도 있습니다. 왜 일부러 오류를 낼까요?

점수에 `150` 이 들어왔다고 합시다. 조용히 받아들이면 평균이 틀립니다. 그리고 **아무도 모릅니다.** 2-1 의 표에서 가장 찾기 어렵다고 한 **논리 오류**가 됩니다. 잘못된 값은 **들어오는 순간** 큰 소리로 막는 것이 낫습니다. 9주차에 `__init__` 에서 값을 검사하던 것과 같은 생각입니다.

````python
def register(name, score):
    if not 0 <= score <= 100:
        raise ValueError(f"점수는 0 ~ 100 사이여야 합니다: {name} = {score}")     # 여기서 멈추고 안내합니다
    return {"name": name, "score": score}

print(register("김민준", 88))

try:
    register("박지호", 150)
except ValueError as e:
    print("등록 실패 →", e)
````

## 2-6. 함수도 값이다 — 일급 객체

1교시 풀이에서 이런 줄이 있었습니다. `max(scores, key=scores.get)`. `scores.get` 뒤에 **괄호가 없습니다.** 함수를 **부른** 것이 아니라 함수 **자체를** 넘긴 것입니다. 5주차의 `sorted(words, key=len)` 도 같았습니다.

**예상해 보기** — `hello` 는 무엇일까요? `hello("민준")` 은요?

````python
def greet(name):
    return f"안녕, {name}"

hello = greet                 # 괄호 없이 — 함수 자체를 변수에 담았습니다 (부르지 않았습니다)

print(hello("민준"))          # 새 이름으로 부릅니다
print(greet)                  # 함수도 값입니다 — 메모리 어딘가의 객체
print(type(greet))
````

함수는 숫자 · 문자열 · 리스트와 **똑같은 자격**을 가진 값입니다. 변수에 담고, 리스트에 넣고, 다른 함수에 **인자로 넘기고**, 함수에서 **돌려줄** 수 있습니다. 이것을 함수가 **일급 객체**(first-class citizen)라고 합니다. 직역하면 ‘일급 시민’, 즉 다른 값과 **같은 권리**를 가진다는 뜻입니다.

비유하면 함수는 **레시피 카드**입니다. 요리를 하지 않고 카드만 건넬 수 있습니다. 받은 사람이 필요할 때 그 카드대로 요리합니다. `greet` 는 카드, `greet("민준")` 은 요리입니다.

<div style="background:#F3F4F6;color:#1F2937;border-left:6px solid #6B7280;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#6B7280;">📎 왜 하필 ‘시민’이라고 부를까 — first-class citizen 이라는 이름의 유래</b><br>이 말은 영국의 컴퓨터과학자 크리스토퍼 스트레이치(Christopher Strachey)가 1967년에 쓴 표현입니다. 당시 언어(ALGOL)에서 숫자는 변수에 넣고, 인자로 넘기고, 돌려받을 수 있었지만 <b>함수는 이름으로 부르는 것만</b> 할 수 있었습니다. 그는 이것을 두고 ‘ALGOL 에서 함수는 <b>이등 시민</b>(second-class citizen)이다’ 라고 적었습니다. 1960년대 미국 민권 운동 시기라 누구나 알던 말이었습니다. 법적으로는 국민이지만 투표 · 거주 같은 권리를 다 누리지 못하는 사람을 가리킵니다.<br>그래서 <b>값으로서의 권리를 모두 가진 것</b>을 일급 시민이라 부릅니다. 권리는 네 가지입니다. ① 변수에 담을 수 있다 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">say = greet</code> &nbsp;② 인자로 넘길 수 있다 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">apply_twice(greet, ...)</code> &nbsp;③ 반환값으로 돌려줄 수 있다 &nbsp;④ 리스트 · 딕셔너리에 담을 수 있다 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">[len, max, sorted]</code>.<br>핵심은 ‘함수가 특별히 대단하다’가 아니라 <b>‘함수를 다른 값과 차별하지 않는다’</b>는 뜻입니다. Python 에서는 함수가 숫자 · 문자열과 똑같은 시민권을 가집니다. 차별이 없으니 숫자를 넣는 자리에 함수를 넣어도 되고, 그래서 다음 절의 데코레이터(함수를 받아 함수를 돌려주는 함수)가 가능합니다.</div>

````python
def shout(text):
    return text.upper() + "!"

def whisper(text):
    return text.lower() + "..."

for style in [shout, whisper, len]:          # 함수들의 리스트 — 내장 함수 len 도 함께
    print(style.__name__, "→", style("Hello World"))
````

````python
def apply_twice(func, x):
    """함수를 받아 두 번 적용한다"""
    return func(func(x))

def add_three(n):
    return n + 3

print(apply_twice(add_three, 10))            # add_three(add_three(10)) = 16
print(apply_twice(shout, "hi"))              # shout(shout("hi"))
````

## 2-7. 데코레이터 — 함수를 포장하는 함수

10주차에 선택 정렬과 병합 정렬의 **실행 시간**을 쟀습니다. 함수가 열 개라면 열 함수 안에 모두 시간 재는 코드를 넣어야 할까요? 함수의 **본래 일**(정렬)과 **곁다리 일**(시간 재기)이 섞여 버립니다.

함수가 값이라면 다른 길이 있습니다. **함수를 받아서, 시간 재는 일을 덧붙인 새 함수를 돌려주는 함수**를 만듭니다. 원래 함수는 건드리지 않고 **포장**만 합니다.

````python
import time

def timed(func):
    """함수를 받아 '실행 시간을 출력하는 함수' 로 포장해서 돌려준다."""
    def wrapper(*args):                                    # *args: 인자를 몇 개든 받아 그대로 넘깁니다
        start = time.perf_counter()
        result = func(*args)                               # 원래 함수를 부르고
        elapsed = time.perf_counter() - start
        print(f"  ⏱ {func.__name__}: {elapsed * 1000:.2f} ms")
        return result                                      # 원래 결과를 그대로 돌려줍니다
    return wrapper                                         # 포장된 함수를 돌려줍니다 — 부르지 않고!

def slow_sum(n):
    total = 0
    for i in range(n):
        total += i
    return total

slow_sum = timed(slow_sum)             # 포장한 함수로 바꿔치기 — 이름은 그대로
print(slow_sum(1_000_000))
````

`timed(slow_sum)` 은 `slow_sum` 을 **부르지 않습니다.** 받아서 `wrapper` 라는 새 함수 안에 넣고, 그 `wrapper` 를 돌려줍니다. 그 뒤로 `slow_sum(...)` 을 부르면 사실은 `wrapper` 가 실행되고, `wrapper` 가 시간을 재며 원래 함수를 부릅니다. 선물을 포장지로 싸듯, **원래 함수 둘레에 한 겹을 덧댄** 것입니다.

마지막 줄 `slow_sum = timed(slow_sum)` 은 너무 자주 쓰여서 Python 이 줄임 표기를 만들었습니다. 함수 정의 **바로 위에** `@timed` 를 적는 것입니다. 이 표기가 **데코레이터**(decorator)이고, `timed` 처럼 함수를 포장해 돌려주는 함수를 데코레이터 함수라고 합니다.

````python
@timed                                  # == 아래 함수를 정의한 뒤 slow_max = timed(slow_max)
def slow_max(items):
    best = items[0]
    for x in items:
        if x > best:
            best = x
    return best

import random
data = [random.randint(1, 1000) for _ in range(300_000)]
print(slow_max(data))
print(max(data))                        # 내장 max 와 답이 같은지 — 시간은 재지 않았습니다
````

`slow_max` 의 본문에는 시간 재는 코드가 **한 줄도 없습니다.** 포장지(`@timed`)를 떼면 평범한 함수로 돌아갑니다. 함수 열 개에 시간을 재려면 열 줄의 `@timed` 면 됩니다.

## 2-8. 타입 힌트 — 읽는 사람을 위한 메모

함수를 쓰려면 **무엇을 넣고 무엇이 나오는지** 알아야 합니다. 1교시의 `read_scores(filename)` — `filename` 은 문자열이겠지요. 돌려주는 것은? 코드를 끝까지 읽어야 딕셔너리임을 압니다. 이것을 함수 머리에 **적어 둘** 수 있습니다.

````python
def read_scores(filename: str) -> dict:                 # filename 은 str, 돌려주는 것은 dict
    """이름,점수 파일을 읽어 {이름: 점수} 딕셔너리로 돌려준다."""
    scores = {}
    with open(filename, encoding="utf-8") as f:
        for line in f:
            name, score = line.strip().split(",")
            scores[name] = int(score)
    return scores

def average(scores: dict) -> float:
    return sum(scores.values()) / len(scores)

print(average(read_scores("scores.txt")))
````

매개변수 뒤의 `: str` 과 괄호 뒤의 `-> dict` 가 **타입 힌트**(type hint)입니다. 힌트라는 이름 그대로, **Python 은 이것을 검사하지 않습니다.**

**예상해 보기** — 힌트와 다른 종류를 넣으면 어떻게 될까요?

````python
def double(n: int) -> int:
    return n * 2

print(double(21))
print(double("ab"))          # 힌트는 int 인데 문자열을 넣었습니다 — 오류가 날까요?
````

오류가 나지 않습니다. 힌트는 **사람이 읽는 메모**이지 검사기가 아닙니다. 그래도 적는 이유는 셋입니다.

1. **6개월 뒤의 나**와 동료가 함수 머리만 보고 쓸 수 있습니다. 본문을 읽지 않아도 됩니다.
2. **편집기와 AI 도구**가 힌트를 읽습니다. 잘못된 종류를 넣으면 실행 전에 밑줄을 그어 주고, 자동완성이 정확해집니다. 12주차 이후 AI 와 함께 코딩할 때 힌트가 있는 코드가 훨씬 잘 통합니다.
3. 요즘 라이브러리의 **문서가 전부 이 모양**입니다. `open(file: str, mode: str = "r", encoding: str | None = None)`. 읽을 줄 알아야 합니다.

| 쓰고 싶은 것 | 힌트 |
|:---|:---|
| 정수 · 실수 · 문자열 · 참/거짓 | `int` `float` `str` `bool` |
| 정수의 리스트 | `list[int]` |
| 문자열 → 정수 딕셔너리 | `dict[str, int]` |
| 돌려주는 것이 없음 | `-> None` |
| 정수 또는 없음(`None`) | `int | None` |

## 2-9. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 3</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">함수 실행 시간 재기 — 10주차 정렬 두 가지 비교</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">10주차의 <b>선택 정렬</b>과 Python 내장 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sorted</code> 의 실행 시간을 비교하려 합니다. 두 함수의 <b>본문은 손대지 않고</b>, 숫자 2,000개짜리 리스트로 각각 몇 ms 가 걸리는지 출력하세요. 함수가 열 개로 늘어도 같은 방법이 통해야 합니다. 함수 머리에는 타입 힌트를 답니다.</div></div>

10주차의 선택 정렬을 가져옵니다. 본문은 그대로입니다.

````python
def selection_sort(items):
    """가장 작은 것을 찾아 앞으로 — 10주차 2-4"""
    result = list(items)
    n = len(result)
    for i in range(n):
        smallest = i
        for j in range(i + 1, n):
            if result[j] < result[smallest]:
                smallest = j
        result[i], result[smallest] = result[smallest], result[i]
    return result

import random
data = [random.randint(1, 100_000) for _ in range(2_000)]
````
