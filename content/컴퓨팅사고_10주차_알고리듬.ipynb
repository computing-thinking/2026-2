## 1-2. 재귀 — 함수가 자기 자신을 부른다

7주차 2-6에서 함수가 다른 함수를 부르면 **호출 스택**에 돌아갈 곳이 쌓인다고 했습니다. 그때 “10주차에는 함수가 **자기 자신**을 부른다”고 예고했습니다. 그것이 **재귀(recursion)** 입니다.

**예상해 보기** — 아래 셀은 무엇을 출력할까요? 출력 순서에 주의하세요.

````python
def countdown(n):
    if n == 0:                  # 멈추는 경우
        print("발사!")
        return
    print(n)
    countdown(n - 1)            # 자기 자신을 — 더 작은 값으로 — 부릅니다

countdown(3)
````

`countdown(3)` 은 3을 출력하고 `countdown(2)` 를 부릅니다. `countdown(2)` 는 2를 출력하고 `countdown(1)` 을, … `countdown(0)` 은 “발사!”를 출력하고 **돌아갑니다.** 그러면 기다리던 `countdown(1)` 이 끝나고, `countdown(2)` 가 끝나고, `countdown(3)` 이 끝납니다.

7주차의 호출 스택 그림이 네 층으로 쌓였다가 위에서부터 걷힙니다.

```
 ┌──────────────┐
 │ countdown(0) │  ← "발사!" 출력 후 return — 가장 나중에 쌓이고 가장 먼저 걷힘
 ├──────────────┤
 │ countdown(1) │
 ├──────────────┤
 │ countdown(2) │
 ├──────────────┤
 │ countdown(3) │  ← 맨 처음 호출. 가장 나중에 끝남
 └──────────────┘
```

이번에는 **돌아온 뒤에** 일을 하는 재귀입니다. 결과를 돌려받아 **쌓아 올립니다.**

````python
def factorial(n):
    """n! = n × (n-1) × … × 1"""
    if n <= 1:                         # 멈추는 경우: 1! = 1
        return 1
    return n * factorial(n - 1)        # n! = n × (n-1)!  — 작은 문제의 답에 한 단계를 더합니다

print(factorial(5))
print(factorial(20))
````

`factorial(5)` 는 “5 × `factorial(4)`”인데 `factorial(4)` 의 값을 아직 모릅니다. 그래서 **기다리며** `factorial(4)` 를 부릅니다. `factorial(4)` 도 `factorial(3)` 을 기다리고 … `factorial(1)` 이 1을 돌려주면 그제야 2 × 1, 3 × 2, 4 × 6, 5 × 24 가 **차례로 계산되며 돌아옵니다.**

수학에서 $n! = n \times (n-1)!$ 이라고 정의하는 것과 정확히 같은 모양입니다. **정의가 곧 코드**입니다. 이것이 재귀의 매력입니다.

## 1-3. 재귀의 두 조건 — 멈추는 경우, 작아지는 문제

재귀 함수에는 반드시 두 부분이 있어야 합니다.

| 조건 | 뜻 | `factorial` 에서 |
|:---|:---|:---|
| **멈추는 경우**(base case) | 더 쪼개지 않고 **바로 답하는** 경우 | `if n <= 1: return 1` |
| **작아지는 문제**(recursive case) | 자기 자신을 **더 작은 입력**으로 부르고, 그 답에 한 단계를 더한다 | `n * factorial(n - 1)` |

둘 중 하나라도 빠지면 어떻게 될까요?

<span style="background:#FFF0EE;color:#FF6B5A;border:1px solid #FF6B5A;padding:2px 10px;border-radius:999px;font-weight:bold;">💥 일부러 오류를 내는 셀 — 멈추는 경우가 없으면?</span>

````python
def forever(n):
    return forever(n - 1)       # 멈추는 경우가 없습니다 — n 이 음수가 되어도 계속 부릅니다

forever(10)
````

`RecursionError: maximum recursion depth exceeded` — “재귀가 너무 깊어졌다”는 안내입니다. 호출 스택이 **끝없이 쌓이다** Python 이 정해 둔 한계(약 1,000층)에서 멈춘 것입니다. 6주차의 무한 반복과 같은 실수입니다. 무한 반복은 영원히 돌지만, 무한 재귀는 메모리가 쌓이므로 Python 이 먼저 끊어 줍니다.

재귀 함수를 쓸 때는 두 가지를 **먼저** 적으세요. ① 가장 작은 입력에서 답은 무엇인가? ② 호출할 때마다 입력이 **정말 작아지는가?** 둘이 확인되면 재귀는 반드시 끝납니다.

<div style="text-align:center;margin:10px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><div style="color:#6B7280;font-size:0.9em;margin-top:4px;">원반 4개의 분해. ①과 ③은 ‘원반 3개 옮기기’ — 같은 문제가 더 작아진 것입니다. 원반 3개를 어떻게 옮기는지는 묻지 않습니다.</div></div>

게임의 **자동으로 풀기** 버튼이 바로 이 세 단계를 되풀이한 것입니다. 세 단계를 코드로 옮기면 끝입니다. 함수 이름은 `hanoi(n, source, target, temp)` — 원반 `n` 개를 `source` 에서 `target` 으로, `temp` 를 중간 단계로 써서 옮깁니다. ①에서는 target 이, ③에서는 source 가 중간 단계가 되므로 **인자의 자리만 바뀝니다.**

````python
def hanoi(n, source, target, temp):
    """원반 n 개를 source 막대기에서 target 막대기로 옮긴다. temp 는 중간 단계."""
    if n == 1:                                   # 멈추는 경우: 원반 하나는 그냥 옮긴다
        print(f"원반 1 을 {source} → {target}")
        return
    hanoi(n - 1, source, temp, target)           # ① 위의 n-1 개를 temp 로 비켜 둔다 (target 이 중간 단계)
    print(f"원반 {n} 을 {source} → {target}")    # ② 가장 큰 원반을 target 으로
    hanoi(n - 1, temp, target, source)           # ③ 비켜 둔 n-1 개를 target 으로 (source 가 중간 단계)

hanoi(3, "source", "target", "temp")
````

출력된 7줄을 위 그림에 그려 가며 따라가 보세요. 규칙을 한 번도 어기지 않습니다. 코드 어디에도 “작은 원반 위에 큰 원반을 올리지 말라”는 검사가 **없는데도** 그렇습니다. 분해가 옳으면 규칙은 저절로 지켜집니다.

`n = 4` 로 바꾸면 15줄이 나옵니다. **코드는 한 글자도 바뀌지 않습니다.** 1,023 줄짜리 설명을 쓰지 않아도 되는 이유입니다.

### 몇 번 옮기나 — 패턴 인식

이동 횟수만 세는 함수로 바꿔 봅니다. 출력 대신 **횟수를 돌려줍니다.**

````python
def hanoi_count(n):
    if n == 1:
        return 1
    return hanoi_count(n - 1) + 1 + hanoi_count(n - 1)     # ① + ② + ③

for n in range(1, 11):
    print(n, hanoi_count(n))
````

1, 3, 7, 15, 31, 63, … 보이나요? **다음 수 = 앞의 수 × 2 + 1**, 즉 $2^n - 1$ 입니다. 작은 경우를 나열해 규칙을 찾는 것이 **패턴 인식**입니다. 규칙을 찾으면 재귀 없이 바로 계산할 수 있습니다.

````python
moves = 2 ** 64 - 1
print(f"원반 64개: {moves:,} 번")
seconds = moves                           # 1초에 한 번 옮긴다면
years = seconds / (60 * 60 * 24 * 365)
print(f"1초에 한 번씩 옮기면 약 {years:,.0f} 년")
````

## 1-5. 재귀와 반복 — 무엇을 쓸까

`factorial` 은 6주차의 `for` 문으로도 쓸 수 있습니다. 둘 다 맞습니다.

````python
def factorial_loop(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial(10), factorial_loop(10))
````

그러면 `hanoi` 를 `for` 문으로 쓸 수 있을까요? 쓸 수는 있지만 **훨씬 어렵습니다.** 하노이 탑의 분해는 “n−1 개 문제 **두 번** + 한 걸음”이라 **가지를 칩니다.** 반복문은 한 줄로 쭉 가는 문제에 맞고, 가지를 치는 문제는 재귀가 자연스럽습니다.

| | 반복문 | 재귀 |
|:---|:---|:---|
| 어울리는 문제 | 한 줄로 쭉 가는 일 — 합계, 최댓값, 목록 훑기 | 가지를 치는 일 — 하노이 탑, 폴더 안의 폴더, 나무 모양 자료 |
| 생각의 방향 | “처음부터 끝까지 하나씩” | “가장 큰 한 조각만 생각하고 나머지는 자기 자신에게” |
| 비용 | 메모리가 거의 들지 않는다 | 호출마다 스택에 쌓인다. 너무 깊으면 `RecursionError` |
| 함정 | 무한 반복 | 멈추는 경우 누락 · 같은 계산 반복(2교시) |

선택 기준은 하나입니다. **문제의 정의가 재귀적이면 재귀, 아니면 반복.** 하노이 탑은 정의 자체가 “n−1 개 문제”를 품고 있으니 재귀입니다.

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 1-4 ~ 1-5 정리</b><br>① 하노이 탑의 열쇠는 <b>가장 큰 원반 하나만 보는 것</b>입니다. 나머지 n−1 개는 같은 문제이므로 함수 자신에게 맡깁니다.<br>② 분해가 옳으면 규칙 검사 없이도 규칙이 지켜집니다. 코드는 짧고, n 이 커져도 바뀌지 않습니다.<br>③ 작은 경우를 나열해 규칙($2^n - 1$)을 찾는 것이 패턴 인식입니다.<br>④ 가지를 치는 문제는 재귀, 한 줄로 가는 문제는 반복.</div>

# 2교시 · 동적 프로그래밍 — 작은 답을 기억해 큰 답을 만들다

<div style="background:#0EA5A4;color:#FFFFFF;padding:20px 26px;border-radius:14px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#FF6B5A;color:#FFFFFF;padding:3px 14px;border-radius:999px;font-weight:bold;">2교시</span>&nbsp;&nbsp;<span style="font-size:1.5em;font-weight:bold;">계단 오르기와 거스름돈</span><span style="float:right;opacity:0.9;">⏱ 약 50분</span><ul style="margin:10px 0 0 0;line-height:1.7;"><li>작은 경우를 나열해 점화식을 찾는다</li><li>같은 계산을 되풀이하는 재귀가 왜 느려지는지 호출 횟수로 확인한다</li><li>메모이제이션(기억하기)과 표 채우기(동적 프로그래밍)로 고친다</li><li>탐욕법이 틀리는 반례를 찾고, 동적 프로그래밍으로 올바른 답을 구한다</li></ul></div>

## 2-1. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 2</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">계단 오르기</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">계단이 n 칸 있습니다. 한 번에 <b>1칸 또는 2칸</b>씩 오를 수 있습니다. 꼭대기까지 오르는 <b>서로 다른 방법</b>은 몇 가지일까요? 예를 들어 3칸이면 (1,1,1), (1,2), (2,1) 세 가지입니다. 50칸이면?</div></div>

손으로 50칸을 세는 것은 불가능합니다. 하노이 탑처럼 **작은 경우부터** 나열합니다.

| 칸 수 n | 방법 | 가짓수 |
|:---:|:---|:---:|
| 1 | (1) | 1 |
| 2 | (1,1), (2) | 2 |
| 3 | (1,1,1), (1,2), (2,1) | 3 |
| 4 | (1,1,1,1), (1,1,2), (1,2,1), (2,1,1), (2,2) | 5 |
| 5 | … | 8 |

1, 2, 3, 5, 8 … **앞의 두 수를 더하면 다음 수**입니다. 왜 그럴까요?

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 분해 — 마지막 한 걸음만 생각한다</b><br>n 칸 꼭대기에 도착하는 <b>마지막 걸음</b>은 둘 중 하나입니다. 1칸을 올라 도착했거나, 2칸을 올라 도착했거나.<br>· 마지막이 1칸이면, 그 전에 <b>n−1 칸</b> 꼭대기에 있었습니다. 거기까지 가는 방법의 수 = f(n−1)<br>· 마지막이 2칸이면, 그 전에 <b>n−2 칸</b> 꼭대기에 있었습니다. 거기까지 가는 방법의 수 = f(n−2)<br>두 경우는 겹치지 않으므로 <b>f(n) = f(n−1) + f(n−2)</b>. 멈추는 경우는 f(1) = 1, f(2) = 2.<br>하노이 탑과 같은 틀입니다. “전체”를 보지 않고 <b>마지막 한 조각</b>만 보고, 나머지는 같은 문제에 맡깁니다. 이렇게 f(n) 을 더 작은 f 로 나타낸 식을 <b>점화식</b>(recurrence relation)이라고 합니다.</div>

점화식이 나왔으니 1교시처럼 재귀로 옮깁니다.

````python
def climb(n):
    if n <= 2:                               # 멈추는 경우: f(1) = 1, f(2) = 2
        return n
    return climb(n - 1) + climb(n - 2)       # 점화식 그대로

for n in range(1, 11):
    print(n, climb(n), end="  |  ")
````

<div style="background:#F3F4F6;color:#1F2937;border-left:6px solid #6B7280;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#6B7280;">📎 피보나치 수열</b><br>1, 1, 2, 3, 5, 8, 13, … 앞의 두 수를 더해 가는 이 수열은 1202년 피보나치(Fibonacci)가 토끼의 번식을 설명하며 소개했습니다. 계단 오르기의 답은 피보나치 수열을 한 칸 민 것입니다. 해바라기 씨의 배열, 나선 조개 껍데기, 주가 분석 도구에도 나타납니다. 오늘은 수열 자체보다 <b>이 점화식을 계산하는 방법</b>이 주제입니다.</div>

## 2-2. 재귀가 느려지는 이유 — 같은 계산을 또 하고 또 한다

`climb(50)` 을 구해 봅시다. 그 전에 작은 값으로 **시간**을 재 봅니다. 7주차에서 쓴 `time` 모듈입니다.

**예상해 보기** — `climb(25)` 와 `climb(30)` 의 시간은 몇 배 차이 날까요? 입력은 1.2배인데.

````python
import time

for n in [20, 25, 30]:
    start = time.time()
    result = climb(n)
    elapsed = time.time() - start
    print(f"climb({n}) = {result:>8}   {elapsed:.3f} 초")
````

입력이 5 늘 때마다 시간이 **약 11배** 늘어납니다. `climb(50)` 은 이 비율로 하면 **몇 시간**, `climb(100)` 은 우주의 나이보다 깁니다. 하노이 탑의 $2^n$ 이 우리 코드 안에서 일어나고 있습니다.

왜일까요? `climb(5)` 가 무엇을 부르는지 그려 봅니다.

```
climb(5)
├─ climb(4)
│  ├─ climb(3)
│  │  ├─ climb(2)
│  │  └─ climb(1)
│  └─ climb(2)
└─ climb(3)              ← climb(3) 을 또 계산합니다. 위에서 이미 했는데.
   ├─ climb(2)
   └─ climb(1)
```

`climb(3)` 이 두 번, `climb(2)` 가 세 번 계산됩니다. `climb(30)` 에서는 `climb(10)` 이 **수천 번** 계산됩니다. 매번 같은 답이 나오는데도요. 호출 횟수를 직접 세어 봅니다.

````python
calls = 0

def climb_counted(n):
    global calls                             # 호출 횟수만 세는 용도 — 7주차에서 피하라고 한 global 을 여기서만 씁니다
    calls += 1
    if n <= 2:
        return n
    return climb_counted(n - 1) + climb_counted(n - 2)

for n in [10, 20, 25]:
    calls = 0
    climb_counted(n)
    print(f"climb({n}): 함수 호출 {calls:,} 번")
````

`climb(25)` 의 답은 121,393 인데 함수는 **15만 번** 불립니다. 서로 다른 입력은 25개뿐입니다. 나머지는 **전부 중복**입니다.

문제의 원인이 보이니 해결책도 보입니다. **한 번 구한 답을 적어 두고, 다시 물으면 적어 둔 것을 돌려준다.**

## 2-3. 메모이제이션 — 한 번 구한 답은 적어 둔다

5주차의 **딕셔너리**가 ‘메모장’이 됩니다. 함수 첫머리에서 메모장을 보고, 있으면 바로 돌려주고, 없으면 계산한 뒤 **적어 두고** 돌려줍니다. 이것을 **메모이제이션(memoization)** 이라고 합니다. ‘memo(메모)’에서 온 말입니다.

````python
memo = {}                                    # {n: climb(n)} — 구한 답을 적어 두는 메모장

def climb_memo(n):
    if n in memo:                            # 이미 구한 적이 있으면
        return memo[n]                       #   적어 둔 답을 바로 돌려준다
    if n <= 2:
        return n
    result = climb_memo(n - 1) + climb_memo(n - 2)
    memo[n] = result                         # 구한 답을 적어 둔다
    return result

start = time.time()
print(climb_memo(50))
print(climb_memo(90))
print(f"{time.time() - start:.4f} 초")
print(len(memo), "개를 적어 두었습니다")
````

몇 시간이 **0.001초**가 되었습니다. 점화식도, 재귀도 그대로입니다. 달라진 것은 딕셔너리 세 줄뿐입니다. `climb(90)` 을 구하는 데 메모장에 적힌 항목은 88개 — 서로 다른 입력의 수만큼만 계산했습니다.

<details><summary><b>📎 더 알아보기 — <code>functools.lru_cache</code>: 메모이제이션을 한 줄로</b></summary>

Python 표준 라이브러리에 이 일을 대신해 주는 도구가 있습니다. 함수 정의 위에 한 줄을 얹으면(이런 것을 데코레이터라고 합니다. 11주차) 인자별로 결과를 자동으로 기억합니다.

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def climb(n):
    if n <= 2:
        return n
    return climb(n - 1) + climb(n - 2)

climb(90)      # 빠릅니다. memo 딕셔너리를 직접 쓰지 않아도 됩니다
```

원리를 아는 것이 먼저이고, 그다음에 도구를 씁니다.

## 2-4. 거꾸로 가기 — 표를 아래에서부터 채운다 (동적 프로그래밍)

메모이제이션은 **큰 문제에서 출발해** 필요한 작은 답을 구하며 내려갑니다(하향식). 반대로 **작은 답부터 차례로** 표에 채워 올라갈 수도 있습니다(상향식). f(1), f(2) 를 알면 f(3) 이 나오고, f(3) 을 알면 f(4) 가 나옵니다. 재귀가 필요 없습니다. 6주차의 `for` 문과 리스트면 됩니다.

````python
def climb_table(n):
    table = [0] * (n + 1)                    # table[i] = i 칸을 오르는 방법의 수
    table[1] = 1
    if n >= 2:
        table[2] = 2
    for i in range(3, n + 1):
        table[i] = table[i - 1] + table[i - 2]    # 점화식 — 이미 채운 두 칸을 읽어 다음 칸을 채운다
    return table[n]

print(climb_table(50))
print(climb_table(90))
````

표를 채우는 이 방식을 **동적 프로그래밍(dynamic programming, DP)** 이라고 합니다. 이름은 1950년대에 벨먼(Bellman)이 붙였는데, ‘동적’이라는 말에 특별한 뜻은 없습니다. 당시 상사가 수학 연구를 싫어해 그럴듯한 이름을 지었다고 벨먼 자신이 회고했습니다. 뜻은 간단합니다. **작은 문제의 답을 표에 적어 두고, 그것으로 큰 문제의 답을 만든다.**

동적 프로그래밍이 통하는 문제는 두 가지 성질이 있습니다.

| 성질 | 뜻 | 계단 오르기에서 |
|:---|:---|:---|
| **겹치는 부분 문제** | 같은 작은 문제가 여러 번 나온다 | `climb(3)` 이 `climb(4)` 와 `climb(5)` 양쪽에서 필요 |
| **최적 부분 구조** | 큰 문제의 답이 작은 문제의 답으로 만들어진다 | f(n) = f(n−1) + f(n−2) |

하노이 탑은 두 번째는 있지만 첫 번째가 없습니다. ①의 `hanoi(n-1, source, temp, …)` 와 ③의 `hanoi(n-1, temp, target, …)` 는 막대기가 달라 **다른 문제**라 적어 둘 것이 없습니다. 그래서 하노이 탑은 $2^n - 1$ 번보다 빨라질 수 없고, 계단 오르기는 n 번이면 됩니다. **같은 재귀라도 겹치는 부분이 있는지**가 운명을 가릅니다.

## 2-5. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 3</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">거스름돈 — 가장 적은 동전으로</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">동전의 종류가 주어졌을 때, 금액을 거슬러 주는 데 필요한 <b>가장 적은 동전 개수</b>는? 예: 동전이 500 · 100 · 50 · 10원일 때 1,260원 → 500 + 500 + 100 + 100 + 50 + 10 = 6개.</div></div>

누구나 아는 방법이 있습니다. **큰 동전부터 되는 만큼** 씁니다. 500원을 2개 쓰고, 남은 260원에 100원을 2개, 50원 1개, 10원 1개. 이렇게 매 순간 가장 좋아 보이는 선택을 하는 것을 **탐욕법(greedy algorithm)** 이라고 합니다.

````python
def change_greedy(coins, amount):
    """큰 동전부터 되는 만큼 쓴다. coins 는 큰 것부터 정렬되어 있어야 한다."""
    count = 0
    for coin in coins:
        count += amount // coin              # 이 동전을 몇 개 쓸 수 있나
        amount = amount % coin               # 남은 금액
    return count

print(change_greedy([500, 100, 50, 10], 1260))
````

6개, 맞습니다. 한국 동전으로는 탐욕법이 **항상** 최적입니다. 그러면 탐욕법이 거스름돈 문제의 답일까요?

**예상해 보기** — 동전이 1 · 3 · 4원이고 6원을 거슬러 줍니다. 탐욕법은 몇 개를 쓸까요? 가장 적은 개수는 몇 개일까요?

````python
print(change_greedy([4, 3, 1], 6))
````

탐욕법은 4 + 1 + 1 = **3개**입니다. 그런데 3 + 3 = **2개**면 됩니다. 탐욕법이 **틀렸습니다.** 4원을 집는 순간이 가장 좋아 보였지만, 그 선택이 뒤를 망쳤습니다.

이것이 오늘 두 번째로 중요한 교훈입니다. **쉬워 보이는 방법이 맞는지는 반례를 찾아 확인해야 합니다.** 반례 하나면 방법 전체가 무너집니다. 한국 동전에서 통했다고 모든 동전에서 통하는 것이 아닙니다.

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 분해 — 마지막에 쓴 동전 하나만 생각한다</b><br>계단 오르기의 ‘마지막 한 걸음’과 똑같습니다. 금액 m 을 만드는 <b>마지막 동전</b>은 1원 · 3원 · 4원 중 하나입니다.<br>· 마지막이 1원이면, 그 전에 <b>m−1 원</b>을 만들어 두었어야 합니다. 그 최소 개수 = f(m−1)<br>· 마지막이 3원이면 f(m−3), 4원이면 f(m−4)<br>셋 중 <b>가장 작은 것 + 1</b>이 f(m) 입니다. 멈추는 경우는 f(0) = 0 (0원은 동전 0개).<br><b>f(m) = min(f(m−1), f(m−3), f(m−4)) + 1</b>. 같은 f(m−3) 이 여러 곳에서 필요하니(겹치는 부분 문제) 표로 채웁니다.</div>

0원부터 차례로 표를 채웁니다. 만들 수 없는 금액(예: 동전이 2원뿐일 때 3원)은 ‘무한대’로 두어 선택되지 않게 합니다.

````python
def min_coins(coins, amount):
    INF = float("inf")                       # 무한대 — '만들 수 없음' 표시
    table = [INF] * (amount + 1)             # table[m] = m 원을 만드는 최소 동전 수
    table[0] = 0                             # 0 원은 동전 0 개
    for m in range(1, amount + 1):
        for coin in coins:                   # 마지막 동전이 coin 이라면
            if coin <= m and table[m - coin] + 1 < table[m]:
                table[m] = table[m - coin] + 1
    return table[amount] if table[amount] != INF else -1     # 만들 수 없으면 -1

print(min_coins([1, 3, 4], 6))               # 2  (3 + 3)
print(min_coins([500, 100, 50, 10], 1260))   # 6
print(min_coins([2], 3))                     # -1
````

표 `table` 을 직접 출력해 보면 `[0, 1, 2, 1, 1, 2, 2]` — 0원 0개, 1원 1개, 2원 2개, 3원 1개, 4원 1개, 5원 2개(1+4), 6원 2개(3+3). **모든 금액**의 답이 한꺼번에 나옵니다. 그 금액 하나만 묻는 탐욕법과 다른 점입니다.

`min_coins` 는 `change_greedy` 보다 느립니다. 금액 × 동전 종류만큼 일합니다. 그러나 **항상 맞습니다.** “빠르지만 가끔 틀리는 방법”과 “조금 느리지만 항상 맞는 방법” 중 무엇을 쓸지는 문제에 따라 정합니다. 그리고 그 판단은 반례를 찾아본 사람만 할 수 있습니다.

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 2-5 정리 — 탐욕법과 동적 프로그래밍</b><br>① <b>탐욕법</b>: 매 순간 가장 좋아 보이는 것을 고른다. 빠르고 단순하지만, 맞는지는 <b>증명하거나 반례를 찾아야</b> 안다.<br>② 거스름돈은 동전 종류에 따라 탐욕법이 틀린다. 반례: 동전 1 · 3 · 4 로 6원.<br>③ <b>동적 프로그래밍</b>: 마지막 한 조각만 생각해 점화식을 세우고, 0부터 표를 채운다. 느리지만 항상 맞다.<br>④ 계단 오르기(경우의 수)와 거스름돈(최솟값)은 다른 문제이지만 <b>생각의 틀은 하나</b>입니다. 이 틀을 익히는 것이 오늘의 목표입니다.</div>

<details><summary><b>💡 정답 예시 보기 — 먼저 스스로 풀어 본 뒤에 열어 보세요</b></summary><pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def grid_paths(rows, cols):
    table = [[0] * cols for _ in range(rows)]      # table[r][c] = (r, c) 칸까지 가는 경로의 수
    for r in range(rows):
        for c in range(cols):
            if r == 0 or c == 0:
                table[r][c] = 1                    # 첫 행 · 첫 열: 한 길뿐
            else:
                table[r][c] = table[r - 1][c] + table[r][c - 1]     # 위에서 + 왼쪽에서
    return table[rows - 1][cols - 1]

## 3-1. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 4</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">쾨니히스베르크의 일곱 다리</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">18세기 프로이센의 도시 쾨니히스베르크(지금의 칼리닌그라드)는 강이 도시를 네 구역으로 나누고, 일곱 개의 다리가 구역들을 잇고 있었습니다. 시민들의 오랜 궁금증: <b>일곱 다리를 모두 정확히 한 번씩만 건너는 산책길</b>이 있을까? 출발점과 도착점은 달라도 됩니다.</div></div>

```
            ┌─────────────────┐
            │     북쪽 땅 (N)    │
            └──┬─────┬─────┬──┘
               │     │     │          다리 7개:
   ┌───────────┴──┐  │  ┌──┴────────┐   N–W 2개, N–E 1개
   │  서쪽 섬 (W)   ├──┼──┤  동쪽 땅 (E) │   W–E 1개
   └───────────┬──┘  │  └──┬────────┘   S–W 2개, S–E 1개
               │     │     │
            ┌──┴─────┴─────┴──┐
            │     남쪽 땅 (S)    │
            └─────────────────┘
```

1736년 오일러(Euler)가 이 문제를 풀었습니다. 다리를 하나하나 건너 보는 대신, 문제를 **다시 그렸습니다.**

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 추상화 — 중요하지 않은 것을 지운다</b><br>산책에서 중요한 것은 무엇일까요? 땅의 <b>모양</b>은 상관없습니다. 다리의 <b>길이</b>도, 강의 <b>폭</b>도 상관없습니다. 중요한 것은 두 가지뿐입니다. <b>어떤 땅들이 있고, 어떤 다리가 어느 땅을 잇는가.</b><br>그래서 땅을 <b>점</b>으로, 다리를 <b>선</b>으로 바꿉니다. 지도가 아니라 점 4개와 선 7개짜리 그림이 됩니다. 이것이 <b>그래프</b>(graph)입니다 — 통계 그래프가 아니라 ‘점과 선의 구조’라는 뜻의 수학 용어입니다. 점은 <b>정점</b>(vertex), 선은 <b>변</b>(edge)이라고 부릅니다.<br>오일러는 그다음 질문을 바꾸었습니다. “어떤 순서로 건너야 하나?”가 아니라 “<b>각 땅에 다리가 몇 개씩 붙어 있나?</b>”</div>

Python 에서 그래프는 5주차의 **딕셔너리**로 표현합니다. 키는 정점, 값은 그 정점에서 선으로 이어진 정점들의 리스트입니다. 다리가 두 개면 이름이 두 번 들어갑니다.

````python
# 쾨니히스베르크: 땅 4 곳, 다리 7 개. 각 다리는 양쪽 땅의 목록에 한 번씩 들어갑니다
bridges = {
    "N": ["W", "W", "E"],          # 북쪽 땅: 서쪽 섬으로 다리 2 개, 동쪽 땅으로 1 개
    "W": ["N", "N", "S", "S", "E"],
    "E": ["N", "W", "S"],
    "S": ["W", "W", "E"],
}

for land, neighbors in bridges.items():
    print(f"{land}: 다리 {len(neighbors)} 개 → {neighbors}")
````

한 정점에 붙은 선의 수를 그 정점의 **차수(degree)** 라고 합니다. N 은 3, W 는 5, E 는 3, S 는 3. **모두 홀수**입니다.

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 패턴 인식 — 왜 차수가 답을 결정하나</b><br>산책 중간에 지나가는 땅을 생각합니다. 그 땅에 <b>들어오는</b> 다리 하나, <b>나가는</b> 다리 하나를 씁니다. 다시 들르면 또 둘. 모든 다리를 한 번씩만 쓰므로, 중간에 지나가는 땅의 다리 수는 <b>반드시 짝수</b>여야 합니다.<br>홀수 다리를 가진 땅은 <b>출발점이나 도착점</b>일 수밖에 없습니다. 출발점과 도착점은 많아야 둘. 그러므로 <b>홀수 차수의 정점이 0개 또는 2개</b>일 때만 산책길이 있습니다.<br>쾨니히스베르크는 홀수 차수 정점이 <b>4개</b>. 산책길은 <b>없습니다.</b> 다리를 하나도 건너 보지 않고 답했습니다.</div>

이 판단을 함수로 씁니다. 그래프 하나를 받아 “모든 선을 한 번씩 지나는 길이 있는가”를 돌려줍니다. 이런 길을 **오일러 경로(Euler path)** 라고 합니다.

````python
def odd_degree_vertices(graph):
    """차수가 홀수인 정점의 목록"""
    return [v for v, neighbors in graph.items() if len(neighbors) % 2 == 1]

def has_euler_path(graph):
    """모든 변을 정확히 한 번씩 지나는 길이 있으면 True (홀수 차수 정점이 0 개 또는 2 개)"""
    return len(odd_degree_vertices(graph)) in (0, 2)

print(odd_degree_vertices(bridges))
print(has_euler_path(bridges))
````

**예상해 보기** — 다리를 하나 더 놓으면 어떻게 될까요? 북쪽 땅과 남쪽 땅 사이에 여덟 번째 다리를 놓습니다. (실제로 1875년에 비슷한 다리가 놓였습니다.)

````python
import copy
bridges8 = copy.deepcopy(bridges)        # 원본을 건드리지 않도록 복사 (5주차: 리스트 · 딕셔너리는 참조)
bridges8["N"].append("S")
bridges8["S"].append("N")

print(odd_degree_vertices(bridges8))
print(has_euler_path(bridges8))
````

홀수 차수 정점이 W 와 E 둘만 남습니다. 이제 산책길이 **있습니다.** W 에서 출발해 E 에서 끝나면 됩니다.

## 3-2. 그래프를 걷기 — 이웃을 따라가며 모두 방문하기

차수만으로 답이 나오는 문제는 드뭅니다. 보통은 그래프를 **직접 걸어야** 합니다. 걷는 방법에는 두 가지가 있는데, 6주차의 **스택과 큐**가 그 둘을 나눕니다.

가장 기본적인 질문부터: “A 에서 출발해 **갈 수 있는 곳**이 어디까지인가?” 이웃을 따라가되, **이미 가 본 곳은 다시 가지 않습니다.**

````python
friends = {                                  # 작은 친구 관계 그래프
    "지민": ["민수", "하늘"],
    "민수": ["지민", "바다"],
    "하늘": ["지민", "바다"],
    "바다": ["민수", "하늘", "솔"],
    "솔":   ["바다"],
    "유리": ["준"],                          # 유리 - 준 은 따로 떨어진 그룹
    "준":   ["유리"],
}

def reachable(graph, start):
    """start 에서 선을 따라 갈 수 있는 모든 정점의 집합"""
    visited = {start}                        # 가 본 곳 — 5주차의 집합(set)
    queue = [start]                          # 앞으로 둘러볼 곳 — 6주차의 큐
    while queue:
        current = queue.pop(0)               # 큐의 맨 앞을 꺼낸다
        for neighbor in graph[current]:
            if neighbor not in visited:      # 처음 보는 곳이면
                visited.add(neighbor)        #   가 본 곳으로 표시하고
                queue.append(neighbor)       #   큐의 맨 뒤에 넣어 나중에 둘러본다
    return visited

print(reachable(friends, "지민"))
print(reachable(friends, "유리"))
````

`visited` 가 없으면 지민 → 민수 → 지민 → 민수 … 를 영원히 돕니다. “가 본 곳 표시”는 그래프를 걸을 때 **빠뜨리면 안 되는 것**입니다. 재귀의 ‘멈추는 경우’와 같은 역할입니다.

큐를 썼으므로 **가까운 곳부터** 둘러봅니다. 지민의 친구(민수, 하늘)를 먼저, 그다음 친구의 친구(바다), 그다음 친구의 친구의 친구(솔). 이렇게 **가까운 순서로 넓혀 가는** 탐색을 **너비 우선 탐색(BFS, breadth-first search)** 이라고 합니다. 큐 대신 스택을 쓰면 한 방향으로 **깊이 파고드는** 깊이 우선 탐색(DFS)이 됩니다. 1교시의 재귀가 바로 스택이므로, 재귀로 쓴 탐색은 DFS 입니다.

가까운 순서로 둘러본다는 성질이 다음 문제의 열쇠입니다.

## 3-3. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 5</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">미로 최단 경로</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">격자 미로가 있습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">#</code> 은 벽, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">.</code> 은 길, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">S</code> 는 출발, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">G</code> 는 목표입니다. 상하좌우 한 칸씩 움직일 때, S 에서 G 까지 <b>가장 짧은 길은 몇 걸음</b>일까요? 갈 수 없으면 −1 입니다.</div></div>

```

. # . # . .
. # . . . #
. . # # . .
# . . . . G
```

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 추상화 — 미로도 그래프다</b><br>미로의 <b>길 칸 하나</b>를 정점으로, <b>이웃한 길 칸 사이</b>를 변으로 보면 미로는 그래프입니다. 쾨니히스베르크에서 땅과 다리가 하던 역할을 칸과 이웃 관계가 합니다. 그래프가 되었으니 3-2의 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">reachable</code> 을 쓸 수 있습니다.<br>그런데 ‘갈 수 있나’가 아니라 ‘<b>몇 걸음</b>인가’를 물었습니다. BFS 는 <b>가까운 곳부터</b> 둘러보므로, 어떤 칸에 <b>처음 도착한 순간</b>의 걸음 수가 곧 <b>최단 거리</b>입니다. 더 짧은 길이 있었다면 그 길로 먼저 도착했을 것이기 때문입니다. 그러니 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">visited</code> 집합 대신 <b>거리를 적는 딕셔너리</b>를 두면 됩니다.</div>

미로는 문자열 리스트로 둡니다. 칸은 `(행, 열)` 튜플입니다. 이웃은 상하좌우 네 방향인데, 격자 밖이거나 벽이면 뺍니다.

````python
maze = [
    "S..#..",
    ".#.#..",
    ".#...#",
    "..##..",
    "#....G",
]

def find(maze, ch):
    """문자 ch 가 있는 칸의 (행, 열)"""
    for r, row in enumerate(maze):
        for c, cell in enumerate(row):
            if cell == ch:
                return (r, c)

def neighbors(maze, cell):
    """상하좌우 이웃 중 격자 안이고 벽이 아닌 칸들"""
    r, c = cell
    result = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:       # 위 · 아래 · 왼쪽 · 오른쪽
        nr, nc = r + dr, c + dc
        if 0 <= nr < len(maze) and 0 <= nc < len(maze[0]) and maze[nr][nc] != "#":
            result.append((nr, nc))
    return result

start, goal = find(maze, "S"), find(maze, "G")
print(start, goal)
print(neighbors(maze, start))
print(neighbors(maze, (2, 2)))
````

`neighbors` 가 그래프의 `graph[current]` 역할을 합니다. 미리 딕셔너리를 만들지 않고 **물을 때마다** 이웃을 계산할 뿐, 하는 일은 같습니다. 이제 3-2의 `reachable` 에 거리만 더합니다.

````python
def shortest_path_length(maze):
    start, goal = find(maze, "S"), find(maze, "G")
    distance = {start: 0}                    # 칸 → S 에서의 걸음 수. '가 본 곳' 표시도 겸합니다
    queue = [start]
    while queue:
        current = queue.pop(0)
        if current == goal:                  # 목표에 처음 도착 — 이것이 최단 거리
            return distance[current]
        for nxt in neighbors(maze, current):
            if nxt not in distance:          # 처음 보는 칸이면
                distance[nxt] = distance[current] + 1      # 한 걸음 더
                queue.append(nxt)
    return -1                                # 큐가 비었는데 못 만났다 — 갈 수 없다

print(shortest_path_length(maze))
````

9걸음입니다. 미로를 손으로 따라가 확인해 보세요. 길이 여러 개인데 BFS 는 그중 가장 짧은 것을 **자동으로** 고릅니다. “모든 길을 다 찾아 비교한다”는 코드는 어디에도 없습니다. 가까운 곳부터 둘러본다는 **순서**가 그 일을 대신합니다.

거리 표 `distance` 를 미로 위에 그려 보면 물결이 퍼지듯 숫자가 번져 가는 것이 보입니다.

````python
def show_distances(maze):
    start = find(maze, "S")
    distance = {start: 0}
    queue = [start]
    while queue:
        current = queue.pop(0)
        for nxt in neighbors(maze, current):
            if nxt not in distance:
                distance[nxt] = distance[current] + 1
                queue.append(nxt)
    for r, row in enumerate(maze):
        line = ""
        for c, cell in enumerate(row):
            if cell == "#":
                line += " ##"
            elif (r, c) in distance:
                line += f"{distance[(r, c)]:3d}"
            else:
                line += "  ."                # 벽에 막혀 갈 수 없는 칸
        print(line)

show_distances(maze)
````

각 칸에 **어디서 왔는지**(`came_from[칸] = 이전 칸`)를 함께 적어 두면, 목표에서 출발점까지 **거꾸로 따라가며** 길을 복원할 수 있습니다. 1교시의 호출 스택이 돌아갈 곳을 기억하던 것과 같은 생각입니다.

```python
while cell is not None:
    path.append(cell)
    cell = came_from[cell]
path.reverse()          # goal → start 순서로 모았으니 뒤집습니다 (6주차 스택!)
```

| **겹치는 부분 문제** | 같은 작은 문제가 여러 번 — 단순 재귀가 지수적으로 느려지는 원인 |
| **메모이제이션** | 딕셔너리에 적어 두기. 재귀는 그대로, 세 줄 추가 |
| **동적 프로그래밍** | 0부터 표 채우기. 재귀 없이, 항상 맞고 빠르다 |
| **탐욕법** | 매 순간 최선 — 빠르지만 반례가 있을 수 있다 |
| **그래프** | `{정점: [이웃들]}`. 차수 = 이웃 수 |
| **BFS** | 큐 + 가 본 곳 표시. 처음 도착 = 최단 거리 |
| **지수적 증가** | $2^n$ — 64개 원반은 우주 나이의 40배. 알고리듬 선택이 컴퓨터 성능보다 중요한 이유 |

<b>이진 탐색 — 반씩 버리기</b><br><b>틀: 분해(절반으로).</b> <b>오름차순으로 정렬된</b> 리스트 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">items</code> 에서 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target</code> 의 위치(인덱스)를 찾아 돌려주는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">binary_search(items, target)</code> 를 쓰세요. 없으면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">-1</code>.<br>🧠 처음부터 하나씩 보지 않습니다. <b>가운데</b> 원소를 봅니다. 같으면 찾았고, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target</code> 이 더 작으면 <b>왼쪽 절반</b>만, 더 크면 <b>오른쪽 절반</b>만 남깁니다. 한 번 볼 때마다 후보가 절반으로 주니, 100만 개에서도 20번이면 끝납니다 — 하노이 탑의 $2^n$ 을 거꾸로 쓴 것입니다.<br>재귀(<code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">low</code>, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">high</code> 를 매개변수로 더한 도우미 함수)와 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">while</code> 반복 중 편한 쪽으로 쓰세요. 단, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">items.index()</code> 와 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">in</code> 은 쓰지 않습니다.<br><code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">binary_search([1, 3, 5, 7, 9, 11], 7)</code> → <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">3</code></div>

<div style="background:#FFFFFF;color:#1F2937;border:2px solid #FF6B5A;padding:12px 18px;border-radius:12px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#FF6B5A;color:#FFFFFF;padding:2px 12px;border-radius:999px;font-weight:bold;">📝 과제 3</span>&nbsp; <b>장애물이 있는 격자 길 찾기 — DP</b><br><b>틀: 점화식 → 표 채우기.</b> 실습 2-2의 격자에 <b>벽</b>이 생겼습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">grid</code> 는 문자열 리스트이고 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">.</code> 은 길, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">#</code> 은 벽입니다. 왼쪽 위에서 오른쪽 아래까지 오른쪽 · 아래로만 가는 경로의 수 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">count_paths(grid)</code> 를 구하세요.<br>🧠 점화식은 실습 2-2와 같습니다. 단 <b>벽 칸의 경로 수는 0</b> 이고, 첫 행 · 첫 열도 벽이 나오면 그 뒤는 0 입니다. 그래서 첫 행 · 첫 열을 따로 1 로 채우지 말고, 모든 칸에서 ‘위 칸 + 왼쪽 칸’(격자 밖은 0)을 더하되 출발 칸만 1 로 두는 편이 단순합니다. 출발 칸이 벽이면 0.</div>

````python
grid = [
    "....",
    ".#..",
    "...#",
    "....",
]

# 여기에 함수를 정의하세요


print(count_paths(grid))
print(count_paths(["...", "...", "..."]))      # 벽이 없으면 실습 2-2 와 같은 6
````

<b>부분집합의 합 — 메모이제이션</b><br><b>틀: 재귀 + 메모이제이션.</b> 양의 정수 리스트 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">nums</code> 의 원소를 <b>여러 번 써도 되게</b> 더해서 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target</code> 을 만들 수 있으면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">True</code> 를 돌려주는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">can_sum(nums, target)</code> 을 쓰세요.<br>🧠 마지막에 더한 수 하나만 봅니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target</code> 을 만들 수 있다 ⇔ 어떤 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">n</code> 에 대해 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target − n</code> 을 만들 수 있다. 멈추는 경우: <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target == 0</code> 이면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">True</code>(아무것도 안 더함), <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target &lt; 0</code> 이면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">False</code>.<br>메모 없이는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">can_sum([7, 14], 300)</code> 이 <b>끝나지 않습니다</b>(같은 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">target</code> 을 수억 번 묻습니다). 딕셔너리 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">memo</code> 에 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">{target: 결과}</code> 를 적어 두세요. 함수 안에서 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">memo = {}</code> 를 만들고 안쪽 도우미 함수가 쓰거나, 과제 2-3처럼 밖에 두어도 됩니다. 단, 밖에 둘 때는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">nums</code> 가 바뀌면 메모를 비워야 합니다 — 안쪽에 두는 편이 안전합니다.<br><code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">can_sum([5, 3, 4, 7], 7)</code> → <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">True</code>, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">can_sum([2, 4], 7)</code> → <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">False</code></div>
