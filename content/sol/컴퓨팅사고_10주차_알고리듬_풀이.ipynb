# 1교시 · 재귀 — 문제를 같은 모양의 작은 문제로 쪼개다

<div style="background:#0EA5A4;color:#FFFFFF;padding:20px 26px;border-radius:14px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#FF6B5A;color:#FFFFFF;padding:3px 14px;border-radius:999px;font-weight:bold;">1교시</span>&nbsp;&nbsp;<span style="font-size:1.5em;font-weight:bold;">재귀와 하노이 탑</span><span style="float:right;opacity:0.9;">⏱ 약 50분</span><ul style="margin:10px 0 0 0;line-height:1.7;"><li>자기 자신을 부르는 함수(재귀)가 어떻게 끝나는지 호출 스택으로 설명한다</li><li>재귀의 두 조건 — 멈추는 경우와 작아지는 문제 — 를 안다</li><li>하노이 탑을 ‘n−1 개 옮기기’라는 같은 모양의 작은 문제로 분해해 푼다</li></ul></div>

## 1-4. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 1</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">하노이 탑 (Tower of Hanoi)</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">막대기 세 개 <b>source · temp · target</b> 이 있습니다. 서로 크기가 다른 원반 n 개가 <b>작은 것이 맨 위, 큰 것이 맨 아래</b>에 오도록 source 막대기에 꽂혀 있습니다. 이 원반들을 <b>target 막대기</b>로 옮기되, 옮긴 뒤에도 <b>같은 순서</b>로 쌓여 있어야 합니다. 조건은 세 가지입니다.<br>1. temp 막대기를 <b>중간 단계</b>로 활용할 수 있습니다.<br>2. 원반은 한 번에 하나씩 옮기며, 어느 막대기에서든 <b>항상 작은 원반이 큰 원반 위에</b> 오도록 옮겨야 합니다.<br>3. 원반 수가 몇 개이든 <b>가장 적은 횟수</b>로 옮겨야 합니다. 원반 3개라면 7번보다 적게는 불가능하고, 7번보다 많으면 낭비입니다.<br>원반이 n 개일 때 최소 몇 번 옮겨야 할까요? 64개라면?</div></div>

<div style="text-align:center;margin:10px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><div style="color:#6B7280;font-size:0.9em;margin-top:4px;">시작 상태와 목표 상태. 원반 번호가 작을수록 작은 원반입니다.</div></div>

원반 3개는 7번, 4개는 15번. 그런데 **방법을 말로 설명**하려면 막힙니다. “먼저 원반 1을 target 으로, 그다음 원반 2를 temp 로, 다시 원반 1을 temp 로 …” 원반이 10개면 이 설명은 1,023 줄이 됩니다.

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 분해 — 가장 큰 원반 하나만 생각한다</b><br>원반 n 개를 source 에서 target 으로 옮기는 일 전체를 보지 말고, <b>맨 아래 가장 큰 원반 하나</b>만 봅니다. 이 원반이 target 으로 가려면, 그 위에 있던 n−1 개가 <b>모두 temp 에</b> 비켜 있어야 합니다. 그러면 일은 세 단계가 됩니다.<br>① 위의 n−1 개를 source 에서 temp 로 옮긴다 (target 을 중간 단계로 씀)<br>② 가장 큰 원반 하나를 source 에서 target 으로 옮긴다<br>③ temp 에 있던 n−1 개를 target 으로 옮긴다 (source 를 중간 단계로 씀)<br>①과 ③은 <b>원반 n−1 개를 옮기는 같은 문제</b>입니다. 막대기 이름만 다릅니다. “n−1 개를 어떻게 옮기는가”는 묻지 않습니다. 그것은 <b>같은 함수가 알아서</b> 합니다. 원반이 1개이면 그냥 옮깁니다 — 멈추는 경우.</div>

<div style="text-align:center;margin:10px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><div style="color:#6B7280;font-size:0.9em;margin-top:4px;">원반 4개의 분해. ①과 ③은 ‘원반 3개 옮기기’ — 같은 문제가 더 작아진 것입니다. 원반 3개를 어떻게 옮기는지는 묻지 않습니다.</div></div>

게임의 **자동으로 풀기** 버튼이 바로 이 세 단계를 되풀이한 것입니다. 세 단계를 코드로 옮기면 끝입니다. 함수 이름은 `hanoi(n, source, target, temp)` — 원반 `n` 개를 `source` 에서 `target` 으로, `temp` 를 중간 단계로 써서 옮깁니다. ①에서는 target 이, ③에서는 source 가 중간 단계가 되므로 **인자의 자리만 바뀝니다.**

````python
def hanoi(n, source, target, temp):
    """원반 n 개를 source 막대기에서 target 막대기로 옮긴다. temp 는 중간 단계."""
    if n == 1:                                   # 멈추는 경우: 원반 하나는 그냥 옮긴다
        print(f"원반 1 을 {source} → {target}")
        return
    hanoi(n - 1, source, temp, target)           # ① 위의 n-1 개를 temp 로 비켜 둔다 (target 이 중간 단계)
    print(f"원반 {n} 을 {source} → {target}")    # ② 가장 큰 원반을 target 으로
    hanoi(n - 1, temp, target, source)           # ③ 비켜 둔 n-1 개를 target 으로 (source 가 중간 단계)

hanoi(3, "source", "target", "temp")
````

출력된 7줄을 위 그림에 그려 가며 따라가 보세요. 규칙을 한 번도 어기지 않습니다. 코드 어디에도 “작은 원반 위에 큰 원반을 올리지 말라”는 검사가 **없는데도** 그렇습니다. 분해가 옳으면 규칙은 저절로 지켜집니다.

`n = 4` 로 바꾸면 15줄이 나옵니다. **코드는 한 글자도 바뀌지 않습니다.** 1,023 줄짜리 설명을 쓰지 않아도 되는 이유입니다.

### 몇 번 옮기나 — 패턴 인식

이동 횟수만 세는 함수로 바꿔 봅니다. 출력 대신 **횟수를 돌려줍니다.**

````python
def hanoi_count(n):
    if n == 1:
        return 1
    return hanoi_count(n - 1) + 1 + hanoi_count(n - 1)     # ① + ② + ③

for n in range(1, 11):
    print(n, hanoi_count(n))
````

1, 3, 7, 15, 31, 63, … 보이나요? **다음 수 = 앞의 수 × 2 + 1**, 즉 $2^n - 1$ 입니다. 작은 경우를 나열해 규칙을 찾는 것이 **패턴 인식**입니다. 규칙을 찾으면 재귀 없이 바로 계산할 수 있습니다.

````python
moves = 2 ** 64 - 1
print(f"원반 64개: {moves:,} 번")
seconds = moves                           # 1초에 한 번 옮긴다면
years = seconds / (60 * 60 * 24 * 365)
print(f"1초에 한 번씩 옮기면 약 {years:,.0f} 년")
````

<b style="color:#FF6B5A;">💡 ✏️ 실습 1-1 정답 예시</b> — 자릿수의 합 — 재귀로

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def sum_digits(n):
    if n &lt; 10:                              # 멈추는 경우: 한 자리면 그 수 자체
        return n
    return n % 10 + sum_digits(n // 10)     # 마지막 자리 + (나머지 자리들의 합)

print(sum_digits(1234))        # 10
print(sum_digits(2 ** 50))     # 76</pre>

<b style="color:#FF6B5A;">💡 ✏️ 실습 1-2 정답 예시</b> — 거듭제곱 — 재귀로

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def power(x, n):
    if n == 0:
        return 1
    return x * power(x, n - 1)

# 도전 — 절반으로 쪼개기 (호출 횟수가 n 번에서 약 log n 번으로)
def power_fast(x, n):
    if n == 0:
        return 1
    if n % 2 == 0:
        half = power_fast(x, n // 2)
        return half * half
    return x * power_fast(x, n - 1)

print(power(2, 10))        # 1024
print(power(1.5, 3))       # 3.375</pre>

<b style="color:#FF6B5A;">💡 ✏️ 실습 1-3 정답 예시</b> — 하노이 탑 — 이동 목록 돌려주기

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def hanoi_moves(n, source, target, temp):
    if n == 1:
        return [(source, target)]
    return (hanoi_moves(n - 1, source, temp, target)      # ① 위의 n-1 개를 temp 로
            + [(source, target)]                          # ② 가장 큰 원반
            + hanoi_moves(n - 1, temp, target, source))   # ③ 비켜 둔 n-1 개를 target 으로

moves = hanoi_moves(3, &quot;A&quot;, &quot;C&quot;, &quot;B&quot;)

# 2교시 · 탐색과 정렬 — 절반을 버리고, 반으로 나누다

<div style="background:#0EA5A4;color:#FFFFFF;padding:20px 26px;border-radius:14px;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#FF6B5A;color:#FFFFFF;padding:3px 14px;border-radius:999px;font-weight:bold;">2교시</span>&nbsp;&nbsp;<span style="font-size:1.5em;font-weight:bold;">숫자 맞히기와 줄 세우기</span><span style="float:right;opacity:0.9;">⏱ 약 50분</span><ul style="margin:10px 0 0 0;line-height:1.7;"><li>하나씩 보는 방법(선형 탐색)과 절반씩 버리는 방법(이진 탐색)의 차이를 횟수로 확인한다</li><li>이진 탐색을 반복과 재귀 두 가지로 쓰고, 무엇을 쓸지 고르는 기준을 세운다</li><li>정렬이 왜 필요한지 알고, 가장 단순한 정렬(선택 정렬)을 손으로 해 본다</li><li>‘반으로 나눠 각각 정렬한 뒤 합치기’(병합 정렬)가 하노이 탑과 같은 재귀임을 안다</li></ul></div>

## 2-1. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 2</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">숫자 맞히기 — 몇 번이면 반드시 맞힐까</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">친구가 1 부터 1,000 사이의 정수 하나를 마음속에 정했습니다. 여러분이 수를 말하면 친구는 <b>“더 커” · “더 작아” · “맞았어”</b> 셋 중 하나로만 답합니다. 친구가 어떤 수를 정했든 <b>반드시</b> 맞히려면 질문이 최대 몇 번 필요할까요? 범위가 1 부터 10억이면?</div></div>

1 부터 차례로 물으면 운이 나쁘면 1,000번입니다. 친구의 답 **“더 커” · “더 작아”** 를 쓰지 않고 있습니다.

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 분해 — 절반을 버린다</b><br><b>가운데</b>를 묻습니다. “500 이야?” 답이 “더 커”이면 1 ~ 500 은 전부 버립니다. “더 작아”이면 500 ~ 1,000 을 버립니다. 어느 답이든 후보가 <b>절반</b>으로 줍니다.<br>1,000 → 500 → 250 → 125 → 62 → 31 → 15 → 7 → 3 → 1. <b>10번</b>이면 반드시 맞힙니다.<br>남은 후보 안에서 또 가운데를 묻습니다 — <b>같은 문제가 절반 크기로</b> 된 것입니다. 하노이 탑과 같은 분해입니다. 후보가 하나이면 그것이 답 — 멈추는 경우.<br>질문 k 번이면 $2^k$ 개까지 가립니다. 10억은 $2^{30}$ 보다 작으니 <b>30번</b>이면 됩니다. 하노이 탑의 $2^n$ 을 거꾸로 쓴 것입니다. 이렇게 정렬된 후보에서 절반씩 버리며 찾는 것을 <b>이진 탐색</b>(binary search)이라고 합니다.</div>

숫자 맞히기를 그대로 코드로 옮깁니다. 후보 범위를 `low` 와 `high` 두 변수로 들고, 가운데 `mid` 를 묻습니다.

````python
def guess(secret, low=1, high=1000):
    """친구가 정한 수 secret 을 low ~ high 안에서 가운데부터 물어 맞힌다. 질문 횟수를 돌려준다."""
    count = 0
    while low <= high:                      # 후보가 남아 있는 동안
        mid = (low + high) // 2             # 가운데를 묻는다
        count += 1
        if mid == secret:
            print(f"{count}번째 질문: {mid} → 맞았어!")
            return count
        elif mid < secret:
            print(f"{count}번째 질문: {mid} → 더 커")
            low = mid + 1                   # 왼쪽 절반을 버린다
        else:
            print(f"{count}번째 질문: {mid} → 더 작아")
            high = mid - 1                  # 오른쪽 절반을 버린다

guess(731)
````

`secret` 을 바꿔 가며 실행해 보세요. 어떤 수도 10번 안에 맞힙니다. 가장 운이 나쁜 수는 어느 것일까요? 1 부터 1,000 까지 **전부** 넣어 보면 됩니다. 출력 없이 횟수만 세는 판을 만듭니다.

````python
def guess_count(secret, low, high):
    count = 0
    while low <= high:
        mid = (low + high) // 2
        count += 1
        if mid == secret:
            return count
        elif mid < secret:
            low = mid + 1
        else:
            high = mid - 1

for n in [10, 100, 1000]:
    worst = max(guess_count(s, 1, n) for s in range(1, n + 1))     # 모든 수를 다 넣어 보고 가장 많은 횟수
    print(f"1 ~ {n:>5,}:   최대 {worst} 번")
````

범위가 **10배** 커질 때 질문은 **3 ~ 4번**만 늘어납니다. 범위가 **2배**가 되면 질문은 정확히 **한 번** 늘어납니다 — 질문 k 번으로 $2^k$ 개까지 가릴 수 있으니, 필요한 횟수는 $2^k \geq n$ 인 가장 작은 $k$, 즉 약 $\log_2 n$ 입니다. 100만이면 20번, 10억이면 30번($2^{30} \approx 10.7$억). 하나씩 묻는 방법(10억 번)과 비교하면 3천만 배 차이입니다. 이것이 ‘절반 버리기’의 힘입니다. (실습 2-1에서 이 횟수를 계산하는 함수를 직접 씁니다.)

## 2-2. 이진 탐색 — 정렬된 목록에서 찾기

숫자 맞히기는 사실 **“정렬된 목록에서 값 찾기”** 입니다. 1 ~ 1,000 이라는 목록이 이미 작은 순으로 놓여 있으니 가운데를 볼 수 있었습니다. 어떤 리스트든 **오름차순으로 정렬되어 있으면** 같은 방법이 통합니다. 이번에는 값이 **몇 번째**에 있는지(인덱스)를 돌려줍니다. 없으면 `-1`.

````python
def binary_search(items, target):
    """오름차순으로 정렬된 items 에서 target 의 인덱스. 없으면 -1."""
    low, high = 0, len(items) - 1           # 아직 후보인 구간
    while low <= high:
        mid = (low + high) // 2
        if items[mid] == target:
            return mid
        elif items[mid] < target:
            low = mid + 1                   # 왼쪽 절반을 버린다
        else:
            high = mid - 1                  # 오른쪽 절반을 버린다
    return -1

scores = [12, 25, 31, 47, 58, 63, 77, 81, 90, 95]
print(binary_search(scores, 63))            # 5
print(binary_search(scores, 12))            # 0
print(binary_search(scores, 50))            # -1
````

**예상해 보기** — 100만 개짜리 정렬된 리스트에서 맨 끝 값을 찾습니다. 하나씩 보는 **선형 탐색**은 몇 번 비교할까요? 이진 탐색은?

````python
def linear_search(items, target):
    """처음부터 하나씩 본다. (인덱스, 비교 횟수)"""
    for i, value in enumerate(items):
        if value == target:
            return i, i + 1
    return -1, len(items)

def binary_search_count(items, target):
    """이진 탐색. (인덱스, 비교 횟수)"""
    low, high, count = 0, len(items) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        count += 1
        if items[mid] == target:
            return mid, count
        elif items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, count

big = list(range(0, 2_000_000, 2))          # 0, 2, 4, …, 1,999,998 — 100만 개, 정렬됨
print("선형 탐색:", linear_search(big, 1_999_998))
print("이진 탐색:", binary_search_count(big, 1_999_998))
print("없는 값  :", binary_search_count(big, 7))
````

100만 번과 20번. 같은 답을 내는데 일의 양이 5만 배 다릅니다. **알고리듬의 선택이 컴퓨터의 속도보다 중요한** 이유입니다.

단, 이진 탐색에는 전제가 하나 있습니다. **정렬되어 있어야 합니다.** 뒤죽박죽인 리스트에 쓰면 어떻게 될까요?

````python
messy = [47, 12, 90, 31, 77, 25]
print(binary_search(messy, 12))             # 12 가 있는데도…
print(messy.index(12))                      # 실제 위치
````

`-1` — 있는 값을 “없다”고 답합니다. 가운데 `31` 보다 `12` 가 작으니 오른쪽을 버렸는데, `12` 는 왼쪽이 아니라 두 번째에 있었습니다. 절반을 버리는 논리는 **순서가 있을 때만** 성립합니다. 그래서 탐색 앞에는 **정렬**이 옵니다. 그것이 문제 3입니다.

## 2-3. 재귀와 반복 — 무엇을 쓸까

`binary_search` 는 `while` 로 썼습니다. 🧠 상자의 “남은 후보 안에서 같은 문제”라는 말 그대로 **재귀로** 쓸 수도 있습니다. 작아지는 문제는 `low`, `high` 가 좁아진 구간, 멈추는 경우는 구간이 비는 것입니다.

````python
def binary_search_rec(items, target, low, high):
    if low > high:                                            # 멈추는 경우: 후보가 없다
        return -1
    mid = (low + high) // 2
    if items[mid] == target:                                  # 멈추는 경우: 찾았다
        return mid
    if items[mid] < target:
        return binary_search_rec(items, target, mid + 1, high)   # 오른쪽 절반 — 같은 문제, 더 작게
    return binary_search_rec(items, target, low, mid - 1)        # 왼쪽 절반

print(binary_search_rec(scores, 63, 0, len(scores) - 1))
print(binary_search_rec(scores, 50, 0, len(scores) - 1))
````

둘 다 맞습니다. 그러면 언제 무엇을 쓸까요?

| | 반복문 | 재귀 |
|:---|:---|:---|
| 어울리는 문제 | 한 줄로 쭉 가는 일 — 합계, 최댓값, 목록 훑기, 이진 탐색 | 가지를 치는 일 — 하노이 탑, 병합 정렬(2-4), 폴더 안의 폴더 |
| 생각의 방향 | “처음부터 끝까지 하나씩” | “가장 큰 한 조각만 생각하고 나머지는 자기 자신에게” |
| 비용 | 메모리가 거의 들지 않는다 | 호출마다 스택에 쌓인다. 너무 깊으면 `RecursionError` |
| 함정 | 무한 반복 · 범위(`low`, `high`) 갱신 실수 | 멈추는 경우 누락 |

이진 탐색은 다음 구간이 **하나**뿐이라(왼쪽 아니면 오른쪽) 반복문이 자연스럽습니다. 하노이 탑은 n−1 개 문제가 **두 번** 나와 가지를 치니 재귀가 자연스럽습니다. **문제의 정의가 가지를 치면 재귀, 한 줄로 가면 반복.**

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 2-1 ~ 2-3 정리 — 이진 탐색</b><br>① 하나씩 보면 n 번, <b>절반씩 버리면</b> $\log_2 n$ 번. 10억 개도 30번.<br>② 가운데를 보고 왼쪽 또는 오른쪽 절반을 버린다. 후보 구간은 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">low</code>, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">high</code> 두 변수로 든다.<br>③ 전제: <b>정렬되어 있어야 한다.</b> 순서가 없으면 절반을 버릴 근거가 없다.<br>④ 반복과 재귀 둘 다 된다. 가지를 치지 않으니 반복이 자연스럽다.</div>

## 2-4. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 3</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">줄 세우기 — 정렬</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">학생 n 명의 키가 뒤죽박죽 목록으로 있습니다. 키가 작은 순으로 줄을 세우려 합니다. 컴퓨터는 한 번에 <b>두 값을 비교</b>하고 <b>자리를 바꾸는</b> 일만 할 수 있습니다. 어떤 순서로 비교하고 바꿔야 할까요? 10명이면 비교를 몇 번 하나요? 100만 명이면?<br>(문제 2의 이진 탐색은 목록이 정렬되어 있어야 했습니다. 정렬은 그 준비 작업입니다.)</div></div>

```
키:  163  178  155  171  160  182  158  169  175  166      ← 10 명
```

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 가장 작은 사람 하나만 생각한다 — 선택 정렬</b><br>줄 전체를 보지 말고 <b>맨 앞자리에 설 사람 하나</b>만 생각합니다. 열 명 중 가장 작은 사람(155)을 찾아 맨 앞 사람과 자리를 바꿉니다. 이제 맨 앞은 끝났습니다. 남은 아홉 명에서 가장 작은 사람을 찾아 두 번째 자리와 바꿉니다. … <b>같은 일이 한 명 줄어든 채</b> 되풀이됩니다. 이것을 <b>선택 정렬</b>(selection sort)이라고 합니다.<br>비교 횟수: 가장 작은 사람을 찾으려면 나머지 전부와 비교해야 하니 9 + 8 + … + 1 = <b>45번</b>. n 명이면 $n(n-1)/2$, 약 $n^2/2$ 번. 100만 명이면 <b>5천억 번</b>입니다.</div>

````python
def selection_sort(items):
    """가장 작은 것을 찾아 앞으로 보내기를 되풀이한다. 새 리스트를 돌려준다. (원본은 그대로)"""
    result = list(items)                              # 복사본을 정렬합니다 (5주차: 리스트는 참조)
    n = len(result)
    for i in range(n - 1):                            # i 번째 자리를 채운다
        smallest = i
        for j in range(i + 1, n):                     # 남은 것 중 가장 작은 것을 찾는다
            if result[j] < result[smallest]:
                smallest = j
        result[i], result[smallest] = result[smallest], result[i]     # 자리를 바꾼다
    return result

heights = [163, 178, 155, 171, 160, 182, 158, 169, 175, 166]
print(selection_sort(heights))
print(heights)                                        # 원본은 그대로
````

비교 횟수를 직접 세어 봅니다. 안쪽 `for` 가 한 번 돌 때마다 비교 한 번입니다.

````python
def selection_sort_count(items):
    result, n, count = list(items), len(items), 0
    for i in range(n - 1):
        smallest = i
        for j in range(i + 1, n):
            count += 1
            if result[j] < result[smallest]:
                smallest = j
        result[i], result[smallest] = result[smallest], result[i]
    return count

import random
for n in [10, 100, 1000]:
    data = [random.randint(140, 200) for _ in range(n)]
    print(f"{n:>5} 명: 비교 {selection_sort_count(data):>8,} 번   (n(n-1)/2 = {n * (n - 1) // 2:,})")
````

10배 많은 사람을 세우는 데 **100배**의 비교가 듭니다. 100만 명이면 5천억 번 — 1초에 1억 번 비교해도 1시간 반입니다. 더 나은 방법이 있을까요? 1교시의 분해를 다시 꺼냅니다.

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 분해 — 반으로 나누어 각각 정렬하고, 합친다 (병합 정렬)</b><br>열 명을 <b>다섯 명씩 두 줄</b>로 나눕니다. 각 줄을 정렬하는 것은 <b>같은 문제가 절반 크기로</b> 된 것이니 같은 함수에 맡깁니다(재귀). 한 명짜리 줄은 이미 정렬되어 있습니다 — 멈추는 경우.<br>남은 일은 <b>정렬된 두 줄을 하나로 합치기</b>뿐입니다. 이것은 쉽습니다. 두 줄의 <b>맨 앞</b>을 비교해 작은 쪽을 꺼내 새 줄에 붙입니다. 한쪽이 비면 다른 쪽을 통째로 붙입니다. n 명이면 비교 n 번 이하.<br>이것을 <b>병합 정렬</b>(merge sort)이라고 합니다. 하노이 탑의 ‘n−1 개 문제 둘 + 한 걸음’과 똑같은 모양입니다. ‘절반 문제 둘 + 합치기’.<br>비교 횟수: 반으로 나누는 층이 $\log_2 n$ 개이고 각 층에서 합치기에 n 번. 합해서 약 $n \log_2 n$. 100만 명이면 <b>2천만 번</b> — 5천억 번의 2만 분의 1입니다.</div>

합치기(`merge`)부터 씁니다. 정렬된 두 리스트를 받아 정렬된 하나를 돌려줍니다.

````python
def merge(left, right):
    """정렬된 두 리스트를 하나의 정렬된 리스트로 합친다. 두 줄의 맨 앞만 비교한다."""
    result = []
    i, j = 0, 0                                   # 각 줄에서 다음에 볼 자리
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:                   # 맨 앞끼리 비교해 작은 쪽을 꺼낸다
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])                       # 한쪽이 비었으면 남은 쪽을 통째로 붙인다
    result.extend(right[j:])
    return result

print(merge([155, 160, 163, 171, 178], [158, 166, 169, 175, 182]))
````

이제 병합 정렬은 세 줄입니다. 하노이 탑의 세 줄과 비교해 보세요.

````python
def merge_sort(items):
    if len(items) <= 1:                           # 멈추는 경우: 0 ~ 1 개는 이미 정렬됨
        return list(items)
    mid = len(items) // 2
    left = merge_sort(items[:mid])                # ① 앞 절반을 정렬한다 — 같은 문제, 절반 크기
    right = merge_sort(items[mid:])               # ② 뒤 절반을 정렬한다
    return merge(left, right)                     # ③ 둘을 합친다

print(merge_sort(heights))
````

**예상해 보기** — 사람 수가 2배가 될 때 선택 정렬과 병합 정렬의 시간은 각각 몇 배가 될까요? Python 에 내장된 `sorted()` 와도 비교합니다.

````python
import time

for n in [1000, 2000, 4000]:
    data = [random.randint(1, 10 ** 6) for _ in range(n)]
    for name, func in [("선택 정렬", selection_sort), ("병합 정렬", merge_sort), ("sorted()", sorted)]:
        start = time.time()
        func(data)
        print(f"{n:>5} 개  {name:<8} {time.time() - start:7.4f} 초")
    print()
````

선택 정렬은 2배에 **약 4배**, 병합 정렬은 **약 2배**로 늘어납니다. 100만 개가 되면 차이는 수만 배로 벌어집니다. 둘 다 같은 답을 내고, 둘 다 몇 줄짜리 코드입니다. 차이는 오직 **문제를 어떻게 쪼갰는가**입니다.

<details><summary><b>📎 더 알아보기 — Python 의 <code>sorted()</code> 는 무엇을 쓰나</b></summary>

`sorted()` 와 `list.sort()` 는 **팀소트(Timsort)** 라는 알고리듬을 씁니다. 병합 정렬을 바탕으로, 이미 정렬된 구간을 찾아 활용하도록 개선한 것입니다. 2002년 Python 개발자 팀 피터스(Tim Peters)가 만들었고, 지금은 Java 와 Android 도 씁니다. C 로 구현되어 있어 우리가 Python 으로 쓴 병합 정렬보다 훨씬 빠르지만, 비교 횟수의 **모양**($n \log n$)은 같습니다.

실무에서는 정렬을 직접 쓰지 않고 `sorted()` 를 씁니다. 오늘 직접 쓴 이유는 “같은 문제를 다르게 쪼개면 일의 양이 얼마나 달라지는가”를 보기 위해서입니다. 정렬 알고리듬은 이 밖에도 버블 정렬(과제 3), 삽입 정렬, 퀵 정렬 등 여러 가지가 있고, 각각 쪼개는 방법이 다릅니다.

</details>

<b style="color:#FF6B5A;">💡 ✏️ 실습 2-1 정답 예시</b> — 몇 번이면 맞힐까 — 절반 버리기 횟수

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def guess_steps(n):
    count = 0
    while n &gt; 0:                 # 후보가 남아 있는 동안
        n = n // 2               # 가운데를 물으면 (최악의 경우) 절반이 남는다
        count += 1
    return count

for n in [1, 2, 10, 100, 1000, 10 ** 6, 10 ** 9]:
    print(f&quot;{n:&gt;13,} 개 → 최대 {guess_steps(n)} 번&quot;)       # 1 2 4 7 10 20 30</pre>

<b style="color:#FF6B5A;">💡 ✏️ 실습 2-2 정답 예시</b> — 이진 탐색 — 직접 쓰기

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def binary_search(items, target):
    low, high = 0, len(items) - 1             # 아직 후보인 구간
    while low &lt;= high:
        mid = (low + high) // 2
        if items[mid] == target:
            return mid
        elif items[mid] &lt; target:
            low = mid + 1                     # 왼쪽 절반을 버린다
        else:
            high = mid - 1                    # 오른쪽 절반을 버린다
    return -1

data = [1, 3, 5, 7, 9, 11]
print(binary_search(data, 7), binary_search(data, 1), binary_search(data, 11), binary_search(data, 4))     # 3 0 5 -1</pre>

<b style="color:#FF6B5A;">💡 ✏️ 실습 2-3 정답 예시</b> — 두 줄 합치기 — merge (도전)

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def merge(left, right):
    result = []
    i, j = 0, 0
    while i &lt; len(left) and j &lt; len(right):
        if left[i] &lt;= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

print(merge([1, 4, 9], [2, 3, 10]))        # [1, 2, 3, 4, 9, 10]
print(merge([], [5, 6]))                   # [5, 6]</pre>

## 3-1. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 4</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">쾨니히스베르크의 일곱 다리</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">18세기 프로이센의 도시 쾨니히스베르크(지금의 칼리닌그라드)는 강이 도시를 네 구역으로 나누고, 일곱 개의 다리가 구역들을 잇고 있었습니다. 시민들의 오랜 궁금증: <b>일곱 다리를 모두 정확히 한 번씩만 건너는 산책길</b>이 있을까? 출발점과 도착점은 달라도 됩니다.</div></div>

<div style="text-align:center;margin:10px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><div style="color:#6B7280;font-size:0.9em;margin-top:4px;">강이 도시를 네 땅으로 나누고, 일곱 다리가 잇습니다. 섬 W 에는 다리가 다섯 개 붙어 있습니다.</div></div>

1736년 오일러(Euler)가 이 문제를 풀었습니다. 다리를 하나하나 건너 보는 대신, 문제를 **다시 그렸습니다.**

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 추상화 — 중요하지 않은 것을 지운다</b><br>산책에서 중요한 것은 무엇일까요? 땅의 <b>모양</b>은 상관없습니다. 다리의 <b>길이</b>도, 강의 <b>폭</b>도 상관없습니다. 중요한 것은 두 가지뿐입니다. <b>어떤 땅들이 있고, 어떤 다리가 어느 땅을 잇는가.</b><br>그래서 땅을 <b>점</b>으로, 다리를 <b>선</b>으로 바꿉니다. 지도가 아니라 점 4개와 선 7개짜리 그림이 됩니다. 이것이 <b>그래프</b>(graph)입니다 — 통계 그래프가 아니라 ‘점과 선의 구조’라는 뜻의 수학 용어입니다. 점은 <b>정점</b>(vertex), 선은 <b>변</b>(edge)이라고 부릅니다.<br>오일러는 그다음 질문을 바꾸었습니다. “어떤 순서로 건너야 하나?”가 아니라 “<b>각 땅에 다리가 몇 개씩 붙어 있나?</b>”</div>

<div style="text-align:center;margin:10px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><div style="color:#6B7280;font-size:0.9em;margin-top:4px;">오일러는 1736년 논문에서 지도를 이렇게 점과 선으로 다시 그려 문제를 정의했습니다 — 그래프 이론의 첫 논문입니다. 두 땅 사이에 다리가 둘이면 선도 둘입니다.</div></div>

Python 에서 그래프는 5주차의 **딕셔너리**로 표현합니다. 키는 정점, 값은 그 정점에서 선으로 이어진 정점들의 리스트입니다. 다리가 두 개면 이름이 두 번 들어갑니다.

````python
# 쾨니히스베르크: 땅 4 곳, 다리 7 개. 각 다리는 양쪽 땅의 목록에 한 번씩 들어갑니다
bridges = {
    "N": ["W", "W", "E"],          # 북쪽 땅: 서쪽 섬으로 다리 2 개, 동쪽 땅으로 1 개
    "W": ["N", "N", "S", "S", "E"],
    "E": ["N", "W", "S"],
    "S": ["W", "W", "E"],
}

for land, neighbors in bridges.items():
    print(f"{land}: 다리 {len(neighbors)} 개 → {neighbors}")
````

한 정점에 붙은 선의 수를 그 정점의 **차수(degree)** 라고 합니다. N 은 3, W 는 5, E 는 3, S 는 3. **모두 홀수**입니다.

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 패턴 인식 — 왜 차수가 답을 결정하나</b><br>산책 중간에 지나가는 땅을 생각합니다. 그 땅에 <b>들어오는</b> 다리 하나, <b>나가는</b> 다리 하나를 씁니다. 다시 들르면 또 둘. 모든 다리를 한 번씩만 쓰므로, 중간에 지나가는 땅의 다리 수는 <b>반드시 짝수</b>여야 합니다.<br>홀수 다리를 가진 땅은 <b>출발점이나 도착점</b>일 수밖에 없습니다. 출발점과 도착점은 많아야 둘. 그러므로 <b>홀수 차수의 정점이 0개 또는 2개</b>일 때만 산책길이 있습니다.<br>쾨니히스베르크는 홀수 차수 정점이 <b>4개</b>. 산책길은 <b>없습니다.</b> 다리를 하나도 건너 보지 않고 답했습니다.</div>

이 판단을 함수로 씁니다. 그래프 하나를 받아 “모든 선을 한 번씩 지나는 길이 있는가”를 돌려줍니다. 이런 길을 **오일러 경로(Euler path)** 라고 합니다.

````python
def odd_degree_vertices(graph):
    """차수가 홀수인 정점의 목록"""
    return [v for v, neighbors in graph.items() if len(neighbors) % 2 == 1]

def has_euler_path(graph):
    """모든 변을 정확히 한 번씩 지나는 길이 있으면 True (홀수 차수 정점이 0 개 또는 2 개)"""
    return len(odd_degree_vertices(graph)) in (0, 2)

print(odd_degree_vertices(bridges))
print(has_euler_path(bridges))
````

**예상해 보기** — 다리를 하나 더 놓으면 어떻게 될까요? 북쪽 땅과 남쪽 땅 사이에 여덟 번째 다리를 놓습니다. (실제로 1875년에 비슷한 다리가 놓였습니다.)

````python
import copy
bridges8 = copy.deepcopy(bridges)        # 원본을 건드리지 않도록 복사 (5주차: 리스트 · 딕셔너리는 참조)
bridges8["N"].append("S")
bridges8["S"].append("N")

print(odd_degree_vertices(bridges8))
print(has_euler_path(bridges8))
````

홀수 차수 정점이 W 와 E 둘만 남습니다. 이제 산책길이 **있습니다.** W 에서 출발해 E 에서 끝나면 됩니다.

<div style="background:#F3F4F6;color:#1F2937;border-left:6px solid #6B7280;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#6B7280;">📎 그래프 이론의 시작, 그리고 지금</b><br>오일러의 이 풀이가 <b>그래프 이론</b>의 출발점입니다. 290년 뒤, 같은 생각이 곳곳에 있습니다. 지하철 노선도(역 = 정점, 선로 = 변), SNS 친구 관계(사람 = 정점, 친구 = 변), 인터넷(컴퓨터 = 정점, 회선 = 변), 분자 구조(원자 = 정점, 결합 = 변). 내비게이션이 길을 찾는 것도, 검색 엔진이 중요한 페이지를 고르는 것도 그래프 위의 계산입니다. <b>복잡한 현실에서 점과 선만 남기는 추상화</b>가 이 모든 것의 첫걸음입니다.</div>

## 3-2. 그래프를 걷기 — 이웃을 따라가며 모두 방문하기

차수만으로 답이 나오는 문제는 드뭅니다. 보통은 그래프를 **직접 걸어야** 합니다. 걷는 방법에는 두 가지가 있는데, 6주차의 **스택과 큐**가 그 둘을 나눕니다.

가장 기본적인 질문부터: “A 에서 출발해 **갈 수 있는 곳**이 어디까지인가?” 아래 `friends` 그래프에서 지민이 갈 수 있는 사람은 누구누구일까요?

````python
friends = {                                  # 작은 친구 관계 그래프
    "지민": ["민수", "하늘"],
    "민수": ["지민", "바다"],
    "하늘": ["지민", "바다"],
    "바다": ["민수", "하늘", "솔"],
    "솔":   ["바다"],
    "유리": ["준"],                          # 유리 - 준 은 따로 떨어진 그룹
    "준":   ["유리"],
}
````

이웃을 따라가되, **이미 가 본 곳은 다시 가지 않습니다.**

````python
def reachable(graph, start):
    """start 에서 선을 따라 갈 수 있는 모든 정점의 집합"""
    visited = {start}                        # 가 본 곳 — 5주차의 집합(set)
    queue = [start]                          # 앞으로 둘러볼 곳 — 6주차의 큐
    while queue:
        current = queue.pop(0)               # 큐의 맨 앞을 꺼낸다
        for neighbor in graph[current]:
            if neighbor not in visited:      # 처음 보는 곳이면
                visited.add(neighbor)        #   가 본 곳으로 표시하고
                queue.append(neighbor)       #   큐의 맨 뒤에 넣어 나중에 둘러본다
    return visited

print(reachable(friends, "지민"))
print(reachable(friends, "유리"))
````

`visited` 가 없으면 지민 → 민수 → 지민 → 민수 … 를 영원히 돕니다. “가 본 곳 표시”는 그래프를 걸을 때 **빠뜨리면 안 되는 것**입니다. 재귀의 ‘멈추는 경우’와 같은 역할입니다.

큐를 썼으므로 **가까운 곳부터** 둘러봅니다. 지민의 친구(민수, 하늘)를 먼저, 그다음 친구의 친구(바다), 그다음 친구의 친구의 친구(솔). 이렇게 **가까운 순서로 넓혀 가는** 탐색을 **너비 우선 탐색(BFS, breadth-first search)** 이라고 합니다. 큐 대신 스택을 쓰면 한 방향으로 **깊이 파고드는** 깊이 우선 탐색(DFS)이 됩니다. 1교시의 재귀가 바로 스택이므로, 재귀로 쓴 탐색은 DFS 입니다.

가까운 순서로 둘러본다는 성질이 다음 문제의 열쇠입니다.

## 3-3. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 5</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">미로 최단 경로</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">격자 미로가 있습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">#</code> 은 벽, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">.</code> 은 길, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">S</code> 는 출발, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">G</code> 는 목표입니다. 상하좌우 한 칸씩 움직일 때, S 에서 G 까지 <b>가장 짧은 길은 몇 걸음</b>일까요? 갈 수 없으면 −1 입니다.</div></div>

```

. # . # . .
. # . . . #
. . # # . .
# . . . . G
```

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 추상화 — 미로도 그래프다</b><br>미로의 <b>길 칸 하나</b>를 정점으로, <b>이웃한 길 칸 사이</b>를 변으로 보면 미로는 그래프입니다. 쾨니히스베르크에서 땅과 다리가 하던 역할을 칸과 이웃 관계가 합니다. 그래프가 되었으니 3-2의 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">reachable</code> 을 쓸 수 있습니다.<br>그런데 ‘갈 수 있나’가 아니라 ‘<b>몇 걸음</b>인가’를 물었습니다. BFS 는 <b>가까운 곳부터</b> 둘러보므로, 어떤 칸에 <b>처음 도착한 순간</b>의 걸음 수가 곧 <b>최단 거리</b>입니다. 더 짧은 길이 있었다면 그 길로 먼저 도착했을 것이기 때문입니다. 그러니 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">visited</code> 집합 대신 <b>거리를 적는 딕셔너리</b>를 두면 됩니다.</div>

미로는 문자열 리스트로 둡니다. 칸은 `(행, 열)` 튜플입니다. 이웃은 상하좌우 네 방향인데, 격자 밖이거나 벽이면 뺍니다.

````python
maze = [
    "S..#..",
    ".#.#..",
    ".#...#",
    "..##..",
    "#....G",
]

def find(maze, ch):
    """문자 ch 가 있는 칸의 (행, 열)"""
    for r, row in enumerate(maze):
        for c, cell in enumerate(row):
            if cell == ch:
                return (r, c)

def neighbors(maze, cell):
    """상하좌우 이웃 중 격자 안이고 벽이 아닌 칸들"""
    r, c = cell
    result = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:       # 위 · 아래 · 왼쪽 · 오른쪽
        nr, nc = r + dr, c + dc
        if 0 <= nr < len(maze) and 0 <= nc < len(maze[0]) and maze[nr][nc] != "#":
            result.append((nr, nc))
    return result

start, goal = find(maze, "S"), find(maze, "G")
print(start, goal)
print(neighbors(maze, start))
print(neighbors(maze, (2, 2)))
````

`neighbors` 가 그래프의 `graph[current]` 역할을 합니다. 미리 딕셔너리를 만들지 않고 **물을 때마다** 이웃을 계산할 뿐, 하는 일은 같습니다. 이제 3-2의 `reachable` 에 거리만 더합니다.

````python
def shortest_path_length(maze):
    start, goal = find(maze, "S"), find(maze, "G")
    distance = {start: 0}                    # 칸 → S 에서의 걸음 수. '가 본 곳' 표시도 겸합니다
    queue = [start]
    while queue:
        current = queue.pop(0)
        if current == goal:                  # 목표에 처음 도착 — 이것이 최단 거리
            return distance[current]
        for nxt in neighbors(maze, current):
            if nxt not in distance:          # 처음 보는 칸이면
                distance[nxt] = distance[current] + 1      # 한 걸음 더
                queue.append(nxt)
    return -1                                # 큐가 비었는데 못 만났다 — 갈 수 없다

print(shortest_path_length(maze))
````

9걸음입니다. 미로를 손으로 따라가 확인해 보세요. 길이 여러 개인데 BFS 는 그중 가장 짧은 것을 **자동으로** 고릅니다. “모든 길을 다 찾아 비교한다”는 코드는 어디에도 없습니다. 가까운 곳부터 둘러본다는 **순서**가 그 일을 대신합니다.

거리 표 `distance` 를 미로 위에 그려 보면 물결이 퍼지듯 숫자가 번져 가는 것이 보입니다.

````python
def show_distances(maze):
    start = find(maze, "S")
    distance = {start: 0}
    queue = [start]
    while queue:
        current = queue.pop(0)
        for nxt in neighbors(maze, current):
            if nxt not in distance:
                distance[nxt] = distance[current] + 1
                queue.append(nxt)
    for r, row in enumerate(maze):
        line = ""
        for c, cell in enumerate(row):
            if cell == "#":
                line += " ##"
            elif (r, c) in distance:
                line += f"{distance[(r, c)]:3d}"
            else:
                line += "  ."                # 벽에 막혀 갈 수 없는 칸
        print(line)

show_distances(maze)
````

<details><summary><b>📎 더 알아보기 — 몇 걸음이 아니라 <code>어느 길</code>인지 알고 싶다면</b></summary>

각 칸에 **어디서 왔는지**(`came_from[칸] = 이전 칸`)를 함께 적어 두면, 목표에서 출발점까지 **거꾸로 따라가며** 길을 복원할 수 있습니다. 1교시의 호출 스택이 돌아갈 곳을 기억하던 것과 같은 생각입니다.

```python
came_from = {start: None}
# BFS 중: came_from[nxt] = current
# 끝난 뒤:
path = []
cell = goal
while cell is not None:
    path.append(cell)
    cell = came_from[cell]
path.reverse()          # goal → start 순서로 모았으니 뒤집습니다 (6주차 스택!)
```

과제 5 에서 직접 해 봅니다. 내비게이션이 길을 **그려 주는** 원리가 이것입니다. 실제 지도는 길마다 길이가 달라 BFS 대신 다익스트라(Dijkstra) 알고리듬을 쓰지만, ‘가까운 곳부터, 어디서 왔는지 기억하며’라는 뼈대는 같습니다.

<pre style="background:#1F2937;color:#F9FAFB;padding:14px 16px;border-radius:10px;font-family:Consolas,monospace;line-height:1.5;white-space:pre-wrap;overflow-wrap:anywhere;box-sizing:border-box;max-width:100%;">def count_reachable(maze):

|:---|:---|
| **재귀** | 멈추는 경우 + 작아지는 문제. 호출 스택에 쌓였다 돌아온다 |

| **병합 정렬** | 반으로 나눠 각각 정렬(재귀)하고 합친다. 약 $n \log_2 n$ 번 |
