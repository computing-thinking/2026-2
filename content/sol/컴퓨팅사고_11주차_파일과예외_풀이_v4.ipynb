## 1-5. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 1</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">성적 파일로 평균 내기</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">학생 10명의 점수가 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">scores.txt</code> 에 <b>한 줄에 한 명씩</b> <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">이름,점수</code> 모양으로 적혀 있습니다. 이 파일을 읽어 <b>평균</b>과 <b>최고점 학생</b>을 구하고, 결과를 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">report.txt</code> 파일에 <b>써서 남기세요.</b> 점수가 100명, 1만 명이 되어도 코드는 바뀌지 않아야 합니다.<br>(점수를 코드에 직접 적지 않습니다. 코드는 파일의 내용을 모릅니다.)</div></div>

먼저 데이터 파일을 만듭니다. 실제로는 누군가 엑셀에서 저장하거나 다른 프로그램이 만들어 둔 파일이겠지만, 오늘은 우리가 직접 씁니다. 아래 셀을 실행하면 `scores.txt` 가 생깁니다.

```python
students = [("김민준", 88), ("이서연", 95), ("박지호", 72), ("최수아", 91), ("정예준", 64),
            ("강하은", 85), ("조시우", 79), ("윤지아", 97), ("장도윤", 58), ("임하린", 83)]

with open("scores.txt", "w", encoding="utf-8") as f:
    for name, score in students:
        f.write(f"{name},{score}\n")

with open("scores.txt", encoding="utf-8") as f:
    print(f.read())
```

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 분해 — 읽기 · 한 줄 처리 · 모으기 · 계산 · 쓰기</b><br>파일 처리 프로그램은 거의 언제나 같은 다섯 단계입니다.<br>① <b>읽기</b>: <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">with open</code> 으로 열고 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">for line in f</code> 로 한 줄씩.<br>② <b>한 줄 처리</b>: <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">strip()</code> 으로 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">\n</code> 을 떼고, <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">split(&quot;,&quot;)</code> 으로 쪼개고, 점수는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">int()</code> 로. 한 줄은 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">이름,점수</code> 두 조각이므로 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">name, score = ...</code> 로 한 번에 받습니다.<br>③ <b>모으기</b>: 이름으로 점수를 찾아야 하니 딕셔너리 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">{이름: 점수}</code> 가 맞습니다.<br>④ <b>계산</b>: 평균은 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sum / len</code>, 최고점 학생은 값이 가장 큰 키.<br>⑤ <b>쓰기</b>: f-string 으로 문자열을 만들어 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">write</code>.<br>①~③을 함수 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">read_scores(filename)</code> 로 묶으면 파일이 바뀌어도, 1만 명이 되어도 코드는 그대로입니다. 7주차의 ‘한 가지 일을 하는 함수’입니다.</div>

```python
def read_scores(filename):
    """이름,점수 파일을 읽어 {이름: 점수} 딕셔너리로 돌려준다."""
    scores = {}
    with open(filename, encoding="utf-8") as f:
        for line in f:
            name, score = line.strip().split(",")     # "김민준,88\n" → "김민준,88" → ["김민준", "88"]
            scores[name] = int(score)                 # 점수는 문자열 → 정수
    return scores

scores = read_scores("scores.txt")
print(scores)
```

```python
average = sum(scores.values()) / len(scores)
best = max(scores, key=scores.get)        # 값(점수)이 가장 큰 키(이름) — key= 에 함수를 넘겼습니다. 2교시에서 다시 봅니다

print(f"학생 수 {len(scores)}명, 평균 {average:.1f}점, 최고점 {best} ({scores[best]}점)")
```

```python
with open("report.txt", "w", encoding="utf-8") as f:
    f.write(f"학생 수: {len(scores)}\n")
    f.write(f"평균: {average:.1f}\n")
    f.write(f"최고점: {best} ({scores[best]})\n")

with open("report.txt", encoding="utf-8") as f:       # 제대로 써졌는지 다시 열어 확인
    print(f.read())
```

`scores.txt` 의 줄을 100개로 늘려도 코드는 한 글자도 바뀌지 않습니다. 코드는 파일의 **내용**을 모르고 **모양**(한 줄에 `이름,점수`)만 압니다. 이것이 데이터를 코드 밖에 두는 이유입니다.

`FileNotFoundError: [Errno 2] No such file or directory: 'score.txt'` — “그런 파일이 없다”는 안내입니다. 그리고 프로그램은 **거기서 멈춥니다.** 뒤에 100줄이 더 있어도 실행되지 않습니다. 파일이 없는 것은 프로그램의 잘못이 아니라 **밖의 세상**의 사정입니다. 밖의 사정으로 프로그램이 멈추지 않게 하는 방법이 2교시의 **예외 처리**입니다.

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 1교시 정리</b><br>① 변수는 메모리에, 파일은 디스크에. 프로그램이 끝나도 남기려면 <b>파일</b>에 씁니다.<br>② 파일 작업은 <b>열기 → 읽기/쓰기 → 닫기</b>. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">with open(이름, 모드, encoding=&quot;utf-8&quot;) as f:</code> 한 줄이 열기와 닫기를 맡습니다.<br>③ 쓰기는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">&quot;w&quot;</code>(새로) · <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">&quot;a&quot;</code>(덧붙여), <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">write</code> 는 문자열만, 줄바꿈은 직접. 읽기는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">read()</code> · <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">for line in f</code> · <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">splitlines()</code>.<br>④ 데이터 파일은 <b>한 줄이 한 건</b>. 한 줄 처리의 기본형은 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">line.strip().split(&quot;,&quot;)</code> 뒤 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">int()</code>.<br>⑤ 코드는 파일의 <b>모양</b>만 알고 <b>내용</b>은 모릅니다. 그래서 데이터가 커져도 코드는 그대로입니다.</div>

이상한 줄이 섞인 데이터 파일을 만듭니다.

```python
messy = """김민준,88
이서연,95
박지호,결석
최수아,91
정예준
강하은,85
조시우,79
윤지아,97
장도윤,58
임하린,83
"""
with open("scores_messy.txt", "w", encoding="utf-8") as f:
    f.write(messy)
print(messy)
```

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 오류의 범위를 작게 — 두 겹의 try</b><br>오류가 날 수 있는 곳이 <b>둘</b>입니다. 파일을 <b>여는</b> 순간(파일이 없다)과, 한 <b>줄을 처리</b>하는 순간(글자, 칸 수). 둘의 대처가 다릅니다. 파일이 없으면 <b>전부 포기</b>하고 빈 딕셔너리, 줄 하나가 이상하면 <b>그 줄만 건너뛰고 계속</b>.<br>그래서 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">try</code> 가 두 겹입니다. 바깥 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">try</code> 는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">open</code> 만 감싸고 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">FileNotFoundError</code> 를 잡습니다. 안쪽 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">try</code> 는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">for</code> 안에서 <b>한 줄의 처리</b>만 감싸고 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">ValueError</code> 를 잡습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">int(&quot;결석&quot;)</code> 도, 조각이 하나뿐인 줄을 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">name, score</code> 둘로 받는 것도 모두 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">ValueError</code> 입니다.<br>안쪽 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">try</code> 가 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">for</code> <b>안에</b> 있어야 한 줄에서 오류가 나도 다음 줄로 넘어갑니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">for</code> 전체를 감싸면 첫 오류에서 반복이 끝납니다.<br>줄 번호는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">enumerate(f, start=1)</code>. 건너뛴 줄은 <b>반드시 출력</b>합니다. 조용히 건너뛰면 여덟 명만 남은 이유를 아무도 모릅니다.</div>

```python
def read_scores_safe(filename):
    """이름,점수 파일을 읽어 {이름: 점수} 로. 파일이 없으면 빈 딕셔너리, 이상한 줄은 알리고 건너뛴다."""
    scores = {}
    try:
        f = open(filename, encoding="utf-8")              # 바깥 try — 여는 순간만 감쌉니다
    except FileNotFoundError:
        print(f"⚠️ 파일이 없습니다: {filename}")
        return scores                                     # 빈 딕셔너리 — 멈추지 않습니다

    with f:                                               # 이미 연 파일도 with 로 감싸 자동으로 닫게 합니다
        for lineno, line in enumerate(f, start=1):
            try:                                          # 안쪽 try — 한 줄의 처리만 감쌉니다
                name, score = line.strip().split(",")
                scores[name] = int(score)
            except ValueError as e:
                print(f"  {lineno}번째 줄 건너뜀 {line.strip()!r} — {e}")
    return scores

scores = read_scores_safe("scores_messy.txt")
print(len(scores), "명:", scores)
```

```python
print(read_scores_safe("없는파일.txt"))        # 멈추지 않고 빈 딕셔너리
print(read_scores_safe("scores.txt"))          # 정상 파일은 1교시와 같은 결과
```

세 번째 줄과 다섯 번째 줄이 **왜** 건너뛰어졌는지 안내문이 알려 줍니다. `int("결석")` 은 “invalid literal”, 쉼표 없는 줄은 “not enough values to unpack”. 둘 다 `ValueError` 라 한 `except` 로 잡혔습니다.

건너뛸지 멈출지는 **데이터의 주인이 정할 일**입니다. 결석을 0점으로 칠지, 빼고 평균을 낼지는 프로그래머가 몰래 정하면 안 됩니다. 그래서 코드는 **건너뛰었다는 사실을 출력**으로 남깁니다. 조용한 프로그램이 가장 위험합니다.

## 2-9. <div style="background:#1F2937;color:#F9FAFB;padding:14px 22px;border-radius:12px;margin:12px 0;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><span style="background:#F59E0B;color:#1F2937;padding:3px 14px;border-radius:999px;font-weight:bold;">문제 3</span>&nbsp;&nbsp;<span style="font-size:1.3em;font-weight:bold;">함수 실행 시간 재기 — 10주차 정렬 두 가지 비교</span><div style="margin-top:8px;opacity:0.92;line-height:1.7;">10주차의 <b>선택 정렬</b>과 Python 내장 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sorted</code> 의 실행 시간을 비교하려 합니다. 두 함수의 <b>본문은 손대지 않고</b>, 숫자 2,000개짜리 리스트로 각각 몇 ms 가 걸리는지 출력하세요. 함수가 열 개로 늘어도 같은 방법이 통해야 합니다. 함수 머리에는 타입 힌트를 답니다.</div></div>

10주차의 선택 정렬을 가져옵니다. 본문은 그대로입니다.

```python
def selection_sort(items):
    """가장 작은 것을 찾아 앞으로 — 10주차 2-4"""
    result = list(items)
    n = len(result)
    for i in range(n):
        smallest = i
        for j in range(i + 1, n):
            if result[j] < result[smallest]:
                smallest = j
        result[i], result[smallest] = result[smallest], result[i]
    return result

import random
data = [random.randint(1, 100_000) for _ in range(2_000)]
```

<div style="background:#FFF7E6;color:#1F2937;border-left:6px solid #F59E0B;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#B45309;">🧠 함수를 값으로 다루면 본문을 열지 않아도 된다</b><br>본문에 손대지 않고 곁다리 일을 붙이는 도구가 데코레이터였습니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">timed</code> 는 함수를 받아 <b>함수를 돌려주므로</b>, 돌려받은 것을 변수에 담으면 그 변수가 곧 ‘시간을 재는 정렬 함수’입니다.<br>우리가 정의한 함수는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">@timed</code> 로, 남이 만든 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">sorted</code> 는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">timed(sorted)</code> 로 — <b>같은 일</b>입니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">@</code> 는 줄임 표기일 뿐이니까요.<br>결과가 같은지는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">==</code> 로 비교합니다. 포장지는 원래 결과를 그대로 돌려주므로 같아야 합니다.<br>타입 힌트: 함수를 받아 함수를 돌려주니 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">Callable -&gt; Callable</code>. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">typing</code> 모듈의 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">Callable</code> 이 ‘부를 수 있는 것’이라는 뜻입니다.</div>

```python
import time
from typing import Callable

def timed(func: Callable) -> Callable:
    """함수를 받아 '실행 시간을 출력하는 함수' 로 포장해서 돌려준다."""
    def wrapper(*args):
        start = time.perf_counter()
        result = func(*args)
        elapsed = time.perf_counter() - start
        print(f"  ⏱ {func.__name__}: {elapsed * 1000:.2f} ms")
        return result
    return wrapper

@timed                                             # 우리가 정의한 함수 — @ 로 포장
def selection_sort(items: list[int]) -> list[int]:
    """가장 작은 것을 찾아 앞으로 — 10주차 2-4"""
    result = list(items)
    n = len(result)
    for i in range(n):
        smallest = i
        for j in range(i + 1, n):
            if result[j] < result[smallest]:
                smallest = j
        result[i], result[smallest] = result[smallest], result[i]
    return result

timed_sorted = timed(sorted)                       # 남이 만든 함수 — 직접 포장

a = selection_sort(data)
b = timed_sorted(data)
print("결과가 같은가?", a == b)
```

선택 정렬은 수십 ~ 수백 ms, 내장 `sorted` 는 1 ms 안팎입니다. 10주차에서 본 $n^2$ 과 $n \log n$ 의 차이이고, 거기에 `sorted` 는 C 언어로 짜여 있다는 차이가 더해졌습니다. 두 함수의 본문에는 시간 재는 코드가 없습니다. 함수가 열 개여도 `@timed` 열 줄, 또는 `timed(함수)` 열 번입니다.

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 2교시 정리</b><br>① <b>예외</b>는 실행 도중 밖의 사정이 코드의 가정과 다를 때 납니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">try</code> · <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">except</code> 로 받아 프로그램이 <b>멈추지 않게</b> 하고, 종류별로 다르게 대처합니다. 범위는 작게, 건너뛴 사실은 출력으로 남깁니다.<br>② <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">raise</code> 로 잘못된 값을 들어오는 순간 막습니다. 조용한 논리 오류보다 시끄러운 예외가 낫습니다.<br>③ 함수는 <b>값</b>입니다(일급 객체). 괄호 없이 쓰면 함수 자체, 괄호를 붙이면 결과. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">key=len</code> 이 그것이었습니다.<br>④ <b>데코레이터</b>는 함수를 받아 포장한 함수를 돌려줍니다. <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">@timed</code> 는 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">f = timed(f)</code>. 본래 일과 곁다리 일을 분리합니다.<br>⑤ <b>타입 힌트</b> <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">def f(x: int) -&gt; str</code> 는 검사하지 않는 메모입니다. 읽는 사람 · 편집기 · AI 도구를 위한 것입니다.</div>

| 개념 | 한 줄 요약 | 기본형 |
|:---|:---|:---|

| **표준 입출력** | 키보드 · 화면도 미리 열려 있는 파일. `print` 는 `sys.stdout` 에 write, `input()` 은 `sys.stdin` 에서 read | `print(x, file=f)` · `sys.stderr` 는 오류 전용 |

| **타입 힌트** | 검사하지 않는 메모. 읽는 사람 · 편집기 · AI 를 위해 | `def f(x: int) -> str:` |
