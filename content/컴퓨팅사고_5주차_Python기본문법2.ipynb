## 1-1. 왜 여러 값을 묶어야 할까?

학생 세 명의 점수를 저장해 봅시다. 지난주 방식대로라면 변수를 세 개 만듭니다.

````python
score1 = 85
score2 = 92
score3 = 78
print(score1, score2, score3)
````

학생이 30명이면 변수 30개, 300명이면 300개가 필요합니다. 평균을 구하는 식도 끝없이 길어집니다.

그래서 **여러 값을 하나의 이름으로 묶어** 담습니다. 이런 자료를 **집합형 자료(container, 컨테이너)** 라고 합니다.

````python
scores = [85, 92, 78]      # 대괄호 [ ] 로 묶은 '리스트'
print(scores)
print(len(scores))         # 몇 개 들어 있나?
````

Python에는 집합형 자료가 다섯 가지 있습니다. 괄호 모양으로 구분합니다.

| 종류 | 모양 | 한 줄 설명 | 비유 |
|:---|:---|:---|:---|
| **리스트** `list` | `[85, 92, 78]` | 순서대로 나열. 고칠 수 있음 | 연필로 쓴 메모 |
| **튜플** `tuple` | `(2007, 3, 15)` | 순서대로 나열. 고칠 수 없음 | 인쇄된 표 |
| **문자열** `str` | `"hello"` | 글자를 순서대로 나열. 고칠 수 없음 | 도장으로 찍은 글자 |
| **딕셔너리** `dict` | `{"name": "김연세"}` | 이름(키)으로 값을 찾음 | 사전 |
| **집합** `set` | `{1, 3, 6}` | 중복 없음. 순서 없음 | 출석부 |

1교시에는 순서가 있는 **리스트 · 튜플 · 문자열**을, 2교시에는 **딕셔너리 · 집합**을 다룹니다.

### 슬라이싱(slicing) — 여러 원소를 잘라 내기

`리스트[시작:끝]` 은 **시작 번호부터 끝 번호 바로 앞까지** 잘라 새 리스트를 만듭니다.
번호를 원소 **사이의 칼금**이라고 생각하면 쉽습니다. `[1:4]` 는 1번 칼금과 4번 칼금 사이, 즉 원소 3개입니다.

````python
numbers = [2, 4, 6, 8, 10]
print(numbers[1:4])      # 1, 2, 3번
print(numbers[:2])       # 시작을 비우면 처음부터
print(numbers[3:])       # 끝을 비우면 마지막까지
print(numbers[:])        # 둘 다 비우면? 예상해 보세요
````

### 원소 바꾸기 · 추가하기 · 삭제하기

리스트는 만든 뒤에도 **원소를 고칠 수 있습니다.**

````python
a = [1, 2, 3, 4, 5]
a[2] = -1          # 2번 원소를 -1 로
print(a)
a[-1] = 0          # 마지막 원소를 0 으로
print(a)
````

````python
colors = ["red", "blue", "green"]
colors.append("white")       # 맨 뒤에 추가
print(colors)
colors.insert(0, "orange")   # 0번 자리에 끼워 넣기
print(colors)
del colors[1]                # 1번 원소 삭제
print(colors)
````

`append()` 와 `insert()` 는 리스트 뒤에 점을 찍어 쓰는 **메서드(method)** 입니다. 4주차의 `"World".upper()` 와 같은 방식입니다.

### 자주 쓰는 문자열 메서드

| 메서드 | 하는 일 | 예 → 결과 |
|:---|:---|:---|
| `strip()` | 양 끝 공백 제거 | `" hi \n".strip()` → `'hi'` |

## 1-5. 고칠 수 있는 자료의 함정

지금까지 본 세 자료 중 **리스트만 고칠 수 있습니다.** 튜플과 문자열은 고칠 수 없습니다.
고칠 수 있는 자료를 **가변(mutable)**, 고칠 수 없는 자료를 **불변(immutable)** 자료라고 합니다.

**예상해 보기** — 4주차에 `b = a` 는 ‘그 순간의 값을 넣는 것’이라고 배웠습니다. 리스트에서도 그럴까요?

````python
a = [1, 2, 3]
b = a            # b 도 같은 리스트를 가리킨다
b.append(4)      # b 쪽에서 리스트를 고치면…
print(a)
print(b)
````

<div style="background:#E6F7F6;color:#1F2937;border-left:6px solid #0EA5A4;padding:12px 18px;border-radius:8px;margin:8px 0;line-height:1.7;box-sizing:border-box;max-width:100%;overflow-wrap:anywhere;width:auto;"><b style="color:#0EA5A4;">🔑 이름표 두 개, 리스트 하나</b><br><code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">b = a</code> 는 리스트를 복사하지 않습니다. <b>같은 리스트에 이름표를 하나 더 붙입니다.</b> 그래서 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">b</code> 로 고치면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">a</code> 로 봐도 바뀌어 있습니다. 숫자나 문자열은 고칠 수 없는 자료라서 이 일이 생기지 않습니다. 따로 쓰고 싶으면 <code style="background:#FFFFFF;color:#0B7C7B;padding:1px 6px;border-radius:4px;border:1px solid #BFE9E7;">b = a.copy()</code> 로 복사하세요.</div>

````python
a = [1, 2, 3]
b = a.copy()     # 복사본을 만들어 b 에 넣는다
b.append(4)
print(a)
print(b)
````

<details><summary><b>📎 더 알아보기 — 변수는 값을 ‘가리킨다’: <code>id()</code> 와 튜플 속 리스트</b></summary>

Python에서 변수는 값이 든 상자라기보다 **값에 붙인 이름표**입니다. `id()` 는 값이 메모리 어디에 있는지 나타내는 번호를 알려 줍니다. 번호가 같으면 같은 값을 가리키는 것입니다.

```python
a = [1, 2, 3]
b = a
c = a.copy()
print(id(a) == id(b))   # True  — 같은 리스트
print(id(a) == id(c))   # False — 다른 리스트
```

**생각해 보기** — 튜플은 고칠 수 없다고 했습니다. 그런데 아래 코드는 오류 없이 실행됩니다.

```python
t = (1, 2, [1, 2])
t[2][0] = 4
print(t)                # (1, 2, [4, 2])
```

튜플이 고정하는 것은 ‘2번 자리에 **이 리스트**가 있다’는 사실입니다. 그 리스트 안의 값까지 막지는 않습니다. `t[2] = [9, 9]` 처럼 다른 리스트로 바꾸려 하면 오류가 납니다.

</details>

## 2-1. 딕셔너리(dict) — 이름으로 찾는 자료

영어 사전에서는 단어로 뜻을 찾습니다. 몇 번째 페이지인지는 몰라도 됩니다.
딕셔너리도 **키(key)** 로 **값(value)** 을 찾습니다. 중괄호 `{ }` 안에 `키: 값` 쌍을 쉼표로 나열합니다.

````python
student = {"name": "김연세", "major": "자율융합계열", "year": 1}
print(student["name"])
print(student["year"])
print(len(student))        # 키-값 쌍의 개수
````

리스트는 `scores[0]` 처럼 **번호**로 찾습니다. 딕셔너리는 `student["name"]` 처럼 **이름**으로 찾습니다. 번호보다 뜻이 잘 드러납니다.

### 값 바꾸기와 새로 추가하기

같은 문법으로 두 가지 일을 합니다. **예상해 보기** — 아래 셀을 실행하면 `student` 는 어떻게 될까요?

````python
student["year"] = 2               # 있는 키 → 값을 바꾼다
student["club"] = "코딩동아리"       # 없는 키 → 새로 추가한다
print(student)
````

<span style="background:#FFF0EE;color:#FF6B5A;border:1px solid #FF6B5A;padding:2px 10px;border-radius:999px;font-weight:bold;">💥 일부러 오류를 내는 셀 — 없는 키로 찾으면?</span>

````python
student["age"]
````

`KeyError` 는 ‘그런 키가 없다’는 뜻입니다. 키가 없을 수도 있을 때는 `get()` 을 씁니다. 오류 대신 `None`(값 없음)이나 내가 정한 값을 돌려줍니다.

````python
print(student.get("age"))
print(student.get("age", "정보 없음"))
````

### 키 · 값 목록 보기와 `in`

````python
country_code = {"America": 1, "Korea": 82, "China": 86, "Japan": 81}
print("Korea" in country_code)     # in 은 '키'가 있는지 확인합니다
print(82 in country_code)          # 값으로는? 예상해 보세요
print(country_code.keys())
print(country_code.values())
print(country_code.items())
````

## 2-2. 집합(set) — 중복 없는 모음

수학의 집합과 같습니다. **같은 값은 한 번만** 들어가고, **순서가 없습니다.**
출석부에 도장을 두 번 찍어도 출석한 사람은 한 명인 것과 같습니다.

````python
dice = [3, 1, 3, 6, 1, 3]       # 주사위를 여섯 번 던진 기록
kinds = set(dice)               # 리스트 → 집합: 중복이 사라집니다
print(kinds)
print(len(kinds))               # 나온 눈의 종류는 몇 가지?
````

````python
a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a & b)      # 교집합 — 둘 다에 있는 것
print(a | b)      # 합집합 — 어느 쪽에든 있는 것
print(a - b)      # 차집합 — a 에만 있는 것
print(3 in a)
````

## 2-4. 순서대로만 실행하면 부족하다 — 제어문

지금까지의 프로그램은 명령문을 **위에서 아래로 한 줄씩 모두** 실행했습니다. 3주차에 본 **순차 구조**입니다.
그런데 실제 문제는 ‘비가 오면 우산을 챙긴다’처럼 **상황에 따라 할 일이 달라집니다.**

3주차 흐름도에서 **마름모**로 그린 부분, 기억나나요? 그것이 **선택 구조**입니다.
이렇게 다른 문장의 실행 순서를 정하는 문장을 **제어문(control statement)** 이라고 합니다.

| 구조 | 흐름도 | Python | 배우는 때 |
|:---|:---|:---|:---|
| 순차 | 네모가 한 줄로 이어짐 | 명령문을 차례로 씀 | 4주차 |
| **선택** | **마름모에서 참/거짓 두 갈래** | **`if` 문** | **오늘** |
| 반복 | 화살표가 위로 되돌아감 | `while` 문, `for` 문 | 6주차 |

## 2-5. 조건식 — 결과가 `True` / `False` 인 식

마름모 안에 적는 질문이 **조건식(Boolean expression)** 입니다. 계산하면 `True` 나 `False` 가 되는 식입니다.
4주차에 배운 **비교 연산자**가 조건식을 만드는 가장 기본 재료입니다.

| 연산자 | 뜻 | 예 (`temperature = 28`) | 결과 |
|:---:|:---|:---|:---:|
| `==` | 같다 | `temperature == 30` | `False` |
| `!=` | 다르다 | `temperature != 30` | `True` |
| `<` `>` | 작다, 크다 | `temperature > 25` | `True` |
| `<=` `>=` | 작거나 같다, 크거나 같다 | `temperature >= 28` | `True` |

````python
temperature = 28
print(temperature == 30)
print(temperature != 30)
print(temperature >= 25)
````

## 2-6. 숫자가 아닌 값의 비교

### 문자열도 비교할 수 있습니다

`==` 는 두 문자열이 **글자 하나까지 똑같은지** 봅니다. 대 · 소문자도 구분합니다.

## 2-7. 논리 연산자 — 조건 여러 개를 묶기

| 연산자 | 뜻 | 참이 되는 경우 |
|:---:|:---|:---|
| `and` | 그리고 | 양쪽 **모두** 참일 때 |
| `or` | 또는 | 양쪽 중 **하나라도** 참일 때 |
| `not` | 아니다 | 참 ↔ 거짓을 뒤집음 |

### 예제 — 나이가 올바른 값인가?

사용자가 입력한 나이가 **0살 이상 120살 이하**일 때만 올바르다고 합시다. **예상해 보기** — `age = 150` 이면 네 줄의 결과는?

````python
age = 150
print(age >= 0 and age <= 120)     # 올바른 나이인가?
print(0 <= age <= 120)             # 같은 뜻, 수학처럼 이어서 쓰기
print(age < 0 or age > 120)        # 잘못된 나이인가?
print(not (0 <= age <= 120))       # 같은 뜻
````

### 연산자 우선순위 — 4주차 표에 `in` 을 더하면

조건식에 연산자가 여러 개 있으면 **위쪽 순위부터** 계산합니다.

| 순위 | 연산자 | 설명 | 같은 순위끼리는 |
|:---:|:---|:---|:---|
| 1 | `( )` | 괄호 | 안쪽 괄호부터 |
| 2 | `**` | 거듭제곱 | **오른쪽 → 왼쪽** |
| 3 | `+x`, `-x` | 부호 | — |
| 4 | `*`, `/`, `//`, `%` | 곱셈 · 나눗셈 · 몫 · 나머지 | 왼쪽 → 오른쪽 |
| 5 | `+`, `-` | 덧셈 · 뺄셈 | 왼쪽 → 오른쪽 |
| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `in`, `not in` | 비교 — 모두 **같은 순위** | **이어 쓰면 `and` 로 묶임** |
| 7 | `not` | 논리 부정 | — |
| 8 | `and` | 그리고 | 왼쪽 → 오른쪽 |
| 9 | `or` | 또는 | 왼쪽 → 오른쪽 |
| 마지막 | `=`, `+=`, `-=` … | 대입 — 연산자가 아니라 명령문입니다. 오른쪽 식을 **모두 계산한 뒤** 넣습니다 | 오른쪽 → 왼쪽 |

- 계산이 **비교보다 먼저**입니다. `x + 1 > 5` 는 `(x + 1) > 5` 입니다.
- 비교가 **논리 연산보다 먼저**입니다. `age >= 0 and age <= 120` 에 괄호가 필요 없는 이유입니다.
- 논리 연산끼리는 **`not` → `and` → `or`** 순서입니다.
- 같은 순위가 이어지면 대부분 **왼쪽부터** 계산합니다. `10 - 3 - 2` 는 `(10 - 3) - 2` 입니다.
- 예외는 두 가지입니다. `**` 는 **오른쪽부터** 계산합니다. 비교는 한쪽부터 계산하지 않고 `and` 로 묶습니다. `3 > 2 > 1` 은 `3 > 2 and 2 > 1` 입니다.

**예상해 보기** — 세 줄의 결과는?

````python
print(True or False and False)      # and 가 먼저: True or (False and False)
print((True or False) and False)    # 괄호가 먼저
print(not 3 > 5)                    # 비교가 먼저: not (3 > 5)
````

**예상해 보기** — 이번에는 같은 순위끼리입니다. 세 줄의 결과는?

````python
print(10 - 3 - 2)       # 왼쪽부터: (10 - 3) - 2
print(2 ** 3 ** 2)      # 오른쪽부터: 2 ** (3 ** 2)
print(3 > 2 > 1)        # 3 > 2 and 2 > 1  — 왼쪽부터라면 True > 1 이 되어 False
````

## 2-8. `if` 문 — 조건이 참일 때만 실행

## 2-9. `if` - `else` — 둘 중 하나

## 2-10. `if` - `elif` - `else` — 여럿 중 하나

갈림길이 셋 이상이면 `elif`(else if 의 줄임말)를 씁니다.

- **위에서부터 차례로** 조건을 검사합니다.
- **처음으로 참이 된 블록 하나만** 실행하고, 나머지는 건너뜁니다.
- 어느 조건도 참이 아니면 `else` 블록을 실행합니다. `else` 는 없어도 됩니다.

````python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:        # 90 미만인 경우만 여기까지 옵니다
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"{score}점은 {grade}학점입니다.")
````

**예상해 보기** — 조건의 **순서**를 바꿨습니다. `95` 점은 몇 학점이 될까요?

````python
score = 95
if score >= 70:
    grade = "C"
elif score >= 80:
    grade = "B"
elif score >= 90:
    grade = "A"
else:
    grade = "F"
print(f"{score}점은 {grade}학점입니다.")
````

**예상해 보기** — `elif` 대신 `if` 를 여러 번 쓰면 무엇이 다를까요?

````python
score = 95
if score >= 90:
    print("A 기준 통과")
if score >= 80:
    print("B 기준 통과")
if score >= 70:
    print("C 기준 통과")
````

`if` 를 따로따로 쓰면 **각각 따로 검사**합니다. 참인 블록이 모두 실행됩니다. **하나만 골라야 할 때**는 `elif` 로 이어 씁니다.

## 2-11. 중첩 `if` 와 `and`

`if` 블록 안에 또 `if` 를 넣을 수 있습니다. 이것을 **중첩 if 문**이라고 합니다.
‘온도가 25도 이상이고, 습도도 70% 이상이면 에어컨을 켠다’를 써 봅시다.

````python
temperature = 28
humidity = 75

if temperature >= 25:
    if humidity >= 70:                 # 한 단계 더 들여씁니다
        print("에어컨을 켭니다.")
print("끝")
````

두 조건이 **모두** 참이어야 할 때는 `and` 로 묶으면 한 줄로 됩니다. 들여쓰기가 얕아져 읽기 쉽습니다.

````python
if temperature >= 25 and humidity >= 70:
    print("에어컨을 켭니다.")
print("끝")
````

## 2-12. 조건식 자리에 숫자 · 문자열 · 리스트를 쓰면?

`if` 뒤에는 `True` / `False` 가 아닌 값도 쓸 수 있습니다. 규칙은 간단합니다.
**‘비어 있거나 0 이면 거짓, 그 밖에는 참’** 입니다.

| 거짓으로 보는 값 | 참으로 보는 값 |
|:---|:---|
| `0`, `0.0`, `""`(빈 문자열), `[]`, `()`, `{}`, `set()`, `None` | 그 밖의 모든 값 — `7`, `"a"`, `[0]` … |

````python
cart = []
if cart:
    print(f"장바구니에 {len(cart)}개가 있습니다.")
else:
    print("장바구니가 비어 있습니다.")
````

`cart` 에 무언가를 넣고 위 셀을 다시 실행해 보세요. 예: `cart = ["연필"]`

# 5주차 정리

| 개념 | 한 줄 요약 |
|:---|:---|
| **리스트** `[ ]` | 순서대로 나열. 번호로 꺼내고 고칠 수 있다. `append` · `insert` · `del` |
| **튜플** `( )` | 리스트와 같지만 고칠 수 없다. `x, y = 1, 2` 도 튜플 |
| **문자열** `" "` | 글자의 나열. 슬라이싱 · `in` 은 되지만 고칠 수 없다 |
| **딕셔너리** `{키: 값}` | 이름(키)으로 찾는다. 없는 키는 `get()` 으로 안전하게 |
| **집합** `{ }` | 중복 없음, 순서 없음. 중복 제거와 교집합 · 합집합 |
| **조건식** | `True` / `False` 가 되는 식. 비교 · `in` · `and` `or` `not` |
| **if 문** | `if` → `elif` → `else`. 콜론과 들여쓰기로 블록을 만든다 |

### 오늘 만난 오류들

| 오류 이름 | 뜻 | 오늘의 예 |
|:---|:---|:---|
| `IndexError` | 번호가 범위를 벗어났다 | `days[7]` |
| `KeyError` | 그런 키가 없다 | `student["age"]` |
| `TypeError` | 그 타입으로는 할 수 없는 일 | 튜플 · 문자열 고치기, `1 < "2"` |
| `SyntaxError` | 문법에 맞지 않는다 | `3 =< 5`, 콜론 빠짐, `if x = 25:` |
| `IndentationError` | 들여쓰기가 맞지 않는다 | `if` 다음 줄을 들여쓰지 않음 |
